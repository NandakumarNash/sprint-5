# 1. Identifying Data Types

Common data types are:

- Numerical Data → Age, Salary, Marks
- Categorical Data → Department, City, Gender
- Boolean Data → True, False
- Date/Time Data → 2026-08-20
- String Data → Name, Address, Product_Name


## AI/ML Example

Study_Hours → Numerical
Attendance → Numerical
Department → Categorical
Passed → Boolean
Exam_Date → Date/Time

- Perform correct calculations
- Apply appropriate visualizations
- Filter data correctly
- Handle dates correctly
- Prepare categorical data
- Avoid conversion errors
- Prepare data correctly for machine learning

Therefore, identifying data types is one of the first steps in data preprocessing and EDA.

# 8. astype()

The astype() function in Pandas is used to convert a column from one data type to another.

It is useful when we already know the correct data type that a column should have.

Common conversions:

String → Integer
Integer → Float
Numeric → Categorical
Integer → String

## AI/ML Example

Suppose a student dataset contains:

Age → "20", "21", "22"
Study_Hours → "5", "6", "8"

These values may be stored as strings.

We can use astype() to convert them into numerical values before giving the data to a machine learning model.

## Business Example

A company may store customer age as text:

Age → "25", "30", "40"

The company can convert Age into integer format to calculate average customer age and perform customer segmentation.

## Real-World Example

An e-commerce application may store product quantity as:

Quantity → "2", "5", "10"

Using astype(int), these values can be converted into numbers so the application can calculate total items sold.

## Why It Matters

astype() is important because it helps us:

- Convert data into the required type
- Perform correct calculations
- Prepare data for analysis
- Prepare features for machine learning
- Reduce type-related errors
- Store data in an appropriate format

Note: astype() should be used carefully because it can produce an error if the value cannot be converted.

Example:

"25" → int works

"abc" → int produces an error

```python

## Python Example

import pandas as pd

data = {
    "Age": ["25", "30", "35"],
    "Experience": [2, 5, 8]
}

df = pd.DataFrame(data)

print(df.dtypes)

df["Age"] = df["Age"].astype(int)
df["Experience"] = df["Experience"].astype(float)

print(df.dtypes)
```

# 9. to_numeric()

pd.to_numeric() is used to convert values into numerical data.

It is especially useful when numbers are stored as strings.

Example:

"25" → 25
"50" → 50
"100" → 100

Unlike astype(), to_numeric() also provides a useful way to handle invalid numerical values using errors="coerce".


## Handling Invalid Values

Suppose the data contains:

Age → ["25", "30", "abc", "40"]

We can use:

df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

The invalid value "abc" becomes NaN.

## AI/ML Example

In a student dataset:

Study_Hours → ["5", "6", "8", "10"]
Marks → ["70", "75", "82", "90"]

We can convert these columns into numerical data before training a machine learning model.

## Business Example

A company may receive sales data from an Excel file:

Sales → ["50000", "75000", "100000"]

Using to_numeric(), the company can convert these values into numbers and calculate total sales and average sales.

## Real-World Example

An online shopping application may receive product prices as text:

Price → ["999", "1499", "2999"]

These values must be converted into numerical data to calculate the total order amount.

## Why It Matters

to_numeric() is important because it helps us:

- Convert numeric strings into numbers
- Perform mathematical calculations
- Handle invalid numerical values
- Detect bad data
- Prepare numerical features for machine learning
- Avoid calculation errors

Therefore, to_numeric() is commonly used during data cleaning and preprocessing.

```python
## Python Example

import pandas as pd

data = {
    "Age": ["25", "30", "35"],
    "Salary": ["30000", "40000", "50000"]
}

df = pd.DataFrame(data)

df["Age"] = pd.to_numeric(df["Age"])
df["Salary"] = pd.to_numeric(df["Salary"])

print(df)
print(df.dtypes)
```

# 11. Handling Invalid Conversions

Invalid conversion happens when a value cannot be converted into the required data type.

Example:

Age → ["25", "30", "abc", "40"]

Here, "abc" cannot be converted into a number.

If we try to convert it directly, Pandas may produce an error.

We can handle invalid values using errors="coerce".



The invalid values become NaN.

## Handling NaN Values

After identifying invalid values, we can handle them using:

df["Age"] = df["Age"].fillna(df["Age"].mean())

This replaces the invalid or missing Age value with the mean age.

## AI/ML Example

Suppose a student dataset contains:

Study_Hours → ["5", "6", "abc", "8"]
Marks → ["70", "75", "invalid", "90"]

Invalid values must be handled before training a machine learning model because ML algorithms generally cannot work directly with invalid text values.

## Business Example

A company receives customer data where:

Purchase_Amount → ["5000", "7500", "unknown", "10000"]

The business must identify and handle "unknown" before calculating total or average purchase amount.

## Real-World Example

An e-commerce application may receive:

Product_Price → ["999", "1499", "N/A", "2999"]

"N/A" is not a valid numerical value.

It can be converted to NaN and then handled appropriately.

## Why It Matters

Handling invalid conversions helps us:

- Prevent conversion errors
- Identify bad data
- Handle missing values
- Improve data quality
- Perform accurate calculations
- Prepare clean data for machine learning

Therefore, invalid values should be detected and handled before performing EDA or machine learning.

```python
## Python Example

import pandas as pd

data = {
    "Age": ["25", "30", "abc", "40"],
    "Salary": ["30000", "40000", "invalid", "50000"]
}

df = pd.DataFrame(data)

df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")

print(df)
```

# 12. Detecting Incorrect Data Types

Incorrect data types occur when a column is stored in a format that does not match the actual meaning of its values.

For example:

Age → "25", "30", "35"

These values are numbers, but if Pandas stores them as object/string, the data type is incorrect for numerical analysis.



## AI/ML Example

In a student dataset:

Study_Hours → "5", "6", "8"
Attendance → "90", "85", "95"

If these columns are stored as strings, a machine learning model may not be able to use them correctly.

They should be converted into numerical data.

## Business Example

A company may have:

Sales → "50000"
Order_Date → "2026-08-20"
Customer_Age → "35"

If these columns are stored as strings, calculations and date analysis may produce incorrect results.

## Real-World Example

In an online shopping dataset:

Price → "999"
Quantity → "2"
Order_Date → "2026-08-20"

Price and Quantity should be numerical, while Order_Date should be Date/Time.

## Why It Matters

Detecting incorrect data types helps us:

- Prevent calculation errors
- Perform correct analysis
- Apply appropriate visualizations
- Handle dates correctly
- Prepare clean ML features
- Avoid unexpected errors

A simple way to detect data types is:

df.dtypes

Therefore, always check the data type of every column before starting EDA and machine learning.

```python
## Python Example

import pandas as pd

data = {
    "Age": ["25", "30", "35"],
    "Salary": [30000, 40000, 50000],
    "Join_Date": ["2026-01-10", "2026-02-15", "2026-03-20"]
}

df = pd.DataFrame(data)

print(df.dtypes)

## The output may show:

## Age → object
## ## Join_Date → object


## Age and Join_Date may need conversion.

#Convert them:

df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Join_Date"] = pd.to_datetime(df["Join_Date"], errors="coerce")

print(df.dtypes)
```

# 13. Why Correct Data Types Are Important

Correct data types are important because Pandas, statistical methods, visualization tools, and machine learning algorithms need data to be in the correct format.

For example:

"100" + "50" = "10050"

Here, the values are strings, so Python performs text concatenation.

But:

100 + 50 = 150

Here, the values are numerical, so Python performs mathematical addition.


## AI/ML Example

A machine learning model may require numerical features.

If:

Study_Hours → "5", "6", "8"

are stored as strings, they should be converted into numerical values before model training.

Correct data types help machine learning algorithms process features correctly.

## Business Example

A company may analyze:

Customer_Type → Categorical
Order_Date → Date/Time

Each value has a different data type and should be handled accordingly.

## Why It Matters

Correct data types help us:

- Perform accurate calculations
- Apply correct statistical methods
- Create appropriate visualizations
- Perform date-based analysis
- Filter and sort data correctly
- Reduce errors
- Improve data quality
- Prepare data for machine learning

Therefore, checking and correcting data types is an important step in data preprocessing and EDA before building a machine learning model.

```python
## Python Example

import pandas as pd

data = {
    "Age": ["25", "30", "35"],
    "Salary": ["30000", "40000", "50000"]
}

df = pd.DataFrame(data)

print(df.dtypes)

df["Age"] = pd.to_numeric(df["Age"])
df["Salary"] = pd.to_numeric(df["Salary"])

print(df.dtypes)

print(df["Salary"].mean())
```
