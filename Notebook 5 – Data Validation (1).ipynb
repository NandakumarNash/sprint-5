# What is Data Validation?

Data Validation is the process of checking whether data is correct, complete, consistent, and follows predefined rules before it is used for analysis, reporting, or machine learning.

## Why is Data Validation Important?

Data validation helps to:

* Identify incorrect values.
* Prevent impossible values.
* Maintain data quality.
* Reduce errors during analysis.
* Improve the reliability of machine learning models.

## Simple Example

Suppose we have student data:

```text
Name    Age
Alice   25
Bob     -5
```

`25` is valid because age cannot be negative.

`-5` is invalid because an age cannot be negative.

## Real-World Example

An online shopping website may validate a customer's pincode before accepting an address. If the pincode does not follow the required format, the system can reject it.

## Business Example

An HR system can validate employee salary. A negative salary should not be accepted.

```text
Salary = 50000  → Valid
Salary = -5000  → Invalid
```

## AI/ML Use Case

Before training a machine learning model, data validation checks whether features such as Age, Salary, Percentage, and Experience contain valid values. Invalid data can negatively affect model training and predictions.


````python
## Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, -5, 30]
})

# Validation rule: Age cannot be negative
df["Age_Valid"] = df["Age"] >= 0

print(df)
````


## How the Output Comes

The validation rule is:

```python
df["Age"] >= 0
```

Python checks every value in the `Age` column individually.

```text
25 >= 0
→ True

-5 >= 0
→ False

30 >= 0
→ True
```

Therefore, Python creates a new column called `Age_Valid`.

```text
Alice    → 25  → True
Bob      → -5  → False
Charlie  → 30  → True
```

`True` means the value passed the validation rule.

`False` means the value failed the validation rule.

So Bob's age is identified as invalid because `-5` is less than `0`.

## Key Point

Data Validation means **checking data against predefined rules before trusting or using that data**.

# Range Validation

Range Validation checks whether a numerical value falls within an allowed minimum and maximum range.

For example:

* Age cannot be negative.
* Salary cannot be negative.
* Percentage must be between 0 and 100.
* Experience may need to be between 0 and 50 years.

## Simple Example

A student's percentage must be between `0` and `100`.

```text
85  → Valid
100 → Valid
0   → Valid
120 → Invalid
-10 → Invalid
```

## Real-World Example

A hospital system may validate a patient's age. An age of `35` is valid, but an age of `-5` is invalid.

## Business Example

An HR system may validate employee salary:

```text
Salary >= 0
```

A salary of `50000` is valid, while `-5000` is invalid.

## AI/ML Use Case

Before training a machine learning model, range validation can identify impossible or incorrect feature values.

For example, if `Percentage` is a feature, a value of `150` may indicate incorrect data entry because percentage should normally be between `0` and `100`.



## How the Output Comes

The important code is:

```python
df["Percentage"].between(0, 100)
```

Python checks each percentage against the range `0` to `100`.

```text
85  → between 0 and 100 → True
120 → between 0 and 100 → False
75  → between 0 and 100 → True
-10 → between 0 and 100 → False
```

Therefore, the `Percentage_Valid` column contains:

```text
85  → True
120 → False
75  → True
-10 → False
```

`True` means the value is within the allowed range.

`False` means the value is outside the allowed range.

## Key Point

Range Validation ensures that numerical values stay within predefined limits.

```python
df["Percentage"].between(0, 100)
```

means:

**"Is each percentage between 0 and 100?"**

# Type Validation

Type Validation checks whether a value has the correct data type expected by the dataset or application.

Common data types include:

* Integer (`int`)
* Decimal (`float`)
* Text (`str`)
* Boolean (`bool`)
* Date (`datetime`)

For example, if `Age` should be a number but the dataset contains `"twenty-five"`, the value fails type validation.

## Simple Example

Suppose we have:

```text id="x8g8wq"
Age
25
30
"twenty-five"
```

The first two values are numbers, but `"twenty-five"` is text.

Therefore:

```text id="e8ckvi"
25            → Valid
30            → Valid
"twenty-five" → Invalid
```

## Real-World Example

A registration form expects a customer's age as a number.

If the user enters:

```text id="1r3n2y"
Age = 25
```

the value is valid.

If the user enters:

```text id="j3kq0z"
Age = "twenty five"
```

the value does not have the expected numerical type.

## Business Example

An HR system expects `Salary` to be numerical because salary calculations are performed on it.

```text id="3j8c3j"
50000      → Correct type
75000.50   → Correct type
"50000"    → Text, may require conversion
```

## AI/ML Use Case

Machine learning algorithms generally require numerical features to perform mathematical calculations. If a numerical column contains unexpected text values, model training can fail or produce incorrect results.



````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, "twenty-five"]
})

# Check whether each value is an integer
df["Age_Type_Valid"] = df["Age"].apply(lambda x: isinstance(x, int))

print(df)
````


## How the Output Comes

The important code is:

```python id="nq4l9j"
isinstance(x, int)
```

`isinstance()` checks whether a value belongs to a particular data type.

Python checks each value:

```text id="p2c3u5"
25
→ Is 25 an integer?
→ Yes
→ True

30
→ Is 30 an integer?
→ Yes
→ True

"twenty-five"
→ Is "twenty-five" an integer?
→ No
→ False
```

Therefore, the `Age_Type_Valid` column becomes:

```text id="x9t8s4"
25            → True
30            → True
"twenty-five" → False
```

`True` means the value has the expected type.

`False` means the value does not have the expected type.

## Another Useful Example

We can check the data type of an entire Pandas column using:

```python id="v0j1dz"
print(df["Age"].dtype)
```

However, because a column can contain mixed types, checking each individual value with `isinstance()` is useful when we specifically want to identify incorrect records.

## Key Point

Type Validation ensures that values have the **expected data type** before they are used for calculations, analysis, or machine learning.


# Format Validation

Format Validation checks whether data follows a required pattern or structure.

It is commonly used for:

* Email addresses
* Phone numbers
* PIN codes
* Employee IDs
* Product codes
* Aadhaar-like identification formats
* Dates stored as text

The value may be present and have the correct data type, but it can still be invalid if its format is wrong.

## Simple Example

Suppose a phone number should contain exactly 10 digits.

```text
9876543210  → Valid
987654321   → Invalid
98765abc10  → Invalid
```

## Real-World Example

A website may ask a customer to enter an email address.

```text
user@gmail.com  → Valid format
usergmail.com   → Invalid format
```

The second value is missing the `@` symbol.

## Business Example

A company may have employee IDs in the format:

```text
EMP001
EMP002
EMP003
```

If someone enters:

```text
ABC123
```

the value does not follow the company's employee ID format.

## AI/ML Use Case

Before using customer data for machine learning, format validation can check fields such as email, phone number, postal code, or product ID. Incorrect formats can indicate data-entry errors and may need to be corrected before model training.

## Python Example

Suppose every employee ID must start with `EMP` followed by exactly 3 digits.

````python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["Alice", "Bob", "Charlie", "David"],
    "Employee_ID": ["EMP001", "EMP002", "ABC003", "EMP04"]
})

# Validate Employee ID format
df["ID_Valid"] = df["Employee_ID"].str.match(r"^EMP\d{3}$")

print(df)
````

## How the Output Comes

The important part is:

```python
r"^EMP\d{3}$"
```

This is a pattern used to describe the required format.

It means:

```text
^       → Start of the value
EMP     → Must contain "EMP"
\d{3}   → Must contain exactly 3 digits
$       → End of the value
```

Now Python checks every Employee ID.

```text
EMP001
→ Starts with EMP
→ Has 3 digits: 001
→ True
```

```text
EMP002
→ Starts with EMP
→ Has 3 digits: 002
→ True
```

```text
ABC003
→ Does not start with EMP
→ False
```

```text
EMP04
→ Starts with EMP
→ Has only 2 digits instead of 3
→ False
```

Therefore:

```text
EMP001 → True
EMP002 → True
ABC003 → False
EMP04  → False
```

`True` means the value follows the required format.

`False` means the value does not follow the required format.

## Key Point

Format Validation checks **whether data follows a required pattern**, not just whether the data exists or has the correct data type.

For example:

```python
df["Employee_ID"].str.match(r"^EMP\d{3}$")
```

means:

**"Check whether every Employee ID starts with `EMP` and is followed by exactly 3 digits."**

# Category Validation

Category Validation checks whether a categorical value belongs to a predefined set of accepted values.

For example, if the accepted Gender categories are:

```text
Male
Female
Other
```

then any value outside this set should be considered invalid.

## Simple Example

Suppose a dataset contains:

```text
Gender
Male
Female
Other
Unknown
```

If the accepted categories are `Male`, `Female`, and `Other`:

```text
Male    → Valid
Female  → Valid
Other   → Valid
Unknown → Invalid
```

## Real-World Example

An online shopping website may allow the following payment methods:

```text
UPI
Card
Cash
```

If a record contains `Bitcoin` as the payment method, it fails category validation.

## Business Example

An HR system may have an employee status field with only these accepted categories:

```text
Active
Inactive
On Leave
```

A value such as `Working` may be invalid because it is not one of the predefined categories.

## AI/ML Use Case

Categorical features are commonly used in machine learning. Category validation ensures that values such as Gender, City, Department, or Product Type contain expected categories before encoding or training the model.



````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Gender": ["Female", "Male", "Other", "Unknown"]
})

# Accepted categories
accepted_categories = ["Male", "Female", "Other"]

# Validate Gender
df["Gender_Valid"] = df["Gender"].isin(accepted_categories)

print(df)
````

## How the Output Comes

The important code is:

```python
df["Gender"].isin(accepted_categories)
```

`.isin()` checks whether each value exists inside the accepted list.

The accepted list is:

```text
["Male", "Female", "Other"]
```

Python checks each Gender value one by one.

```text
Female
→ Is Female in the accepted list?
→ Yes
→ True
```

```text
Male
→ Is Male in the accepted list?
→ Yes
→ True
```

```text
Other
→ Is Other in the accepted list?
→ Yes
→ True
```

```text
Unknown
→ Is Unknown in the accepted list?
→ No
→ False
```

Therefore, the new column becomes:

```text
Female  → True
Male    → True
Other   → True
Unknown → False
```

`True` means the category is accepted.

`False` means the category is not accepted.

## Key Point

Category Validation ensures that categorical values belong to a predefined set of allowed values.

```python
df["Gender"].isin(["Male", "Female", "Other"])
```

means:

**"Check whether each Gender value belongs to the accepted categories."**

# Null Validation

Null Validation checks whether required data is missing or empty.

A null value means that a value is not available or has not been provided.

For example:

```text id="5b0drt"
Name    Age
Alice   25
Bob     NaN
Charlie 30
```

Here, Bob's Age is missing.

## Why is Null Validation Important?

Missing values can cause problems during:

* Data analysis
* Calculations
* Reporting
* Database operations
* Machine learning

Before using a dataset, we should identify which columns contain missing values and decide how to handle them.

## Simple Example

Suppose every employee must have an Age.

```text id="d8t6fr"
Alice    25     → Valid
Bob      NaN    → Invalid
Charlie  30     → Valid
```

The `NaN` value indicates that Bob's age is missing.

## Real-World Example

An online registration form may require a customer's email address.

If the email field is empty, the system should not accept the registration.

```text id="2b3i2x"
Email = "user@gmail.com" → Valid
Email = missing          → Invalid
```

## Business Example

An HR system may require every employee to have an Employee ID.

If an employee record has no Employee ID, the record should be flagged for correction.

## AI/ML Use Case

Machine learning algorithms often cannot directly work with missing values. Null validation helps identify missing values before preprocessing and model training.

The missing values can then be:

* Filled using an appropriate method.
* Removed if appropriate.
* Handled using a model-specific technique.


````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, None, 30]
})

# Check whether Age is NOT null
df["Age_Valid"] = df["Age"].notna()

print(df)
````



## How the Output Comes

The important code is:

```python id="5iyv44"
df["Age"].notna()
```

`.notna()` checks whether each value is present.

Python checks the values one by one.

```text id="2u4ncs"
25
→ Is the value present?
→ Yes
→ True
```

```text id="6m4jiz"
NaN
→ Is the value present?
→ No
→ False
```

```text id="1m7o8r"
30
→ Is the value present?
→ Yes
→ True
```

Therefore:

```text id="9fymt3"
25   → True
NaN  → False
30   → True
```

`True` means the value is present.

`False` means the value is missing.

## Finding the Missing Records

We can use the validation column to find invalid records:

```python id="xy8brn"
print(df[df["Age_Valid"] == False])
```

Output:

```text id="8a2h1j"
  Name  Age  Age_Valid
1  Bob  NaN      False
```

This happens because only Bob has a missing Age value.

## Another Method

We can directly check for missing values using:

```python id="b51q5n"
df["Age"].isna()
```

This gives:

```text id="x9x4c7"
False
True
False
```

Here, `True` means **the value is missing**, which is the opposite meaning of `.notna()`.

## Key Point

```python id="1b2t6g"
.notna()
```

means:

**"Is the value present?"**

while:

```python id="0j5n9m"
.isna()
```

means:

**"Is the value missing?"**


# Business Rule Validation

Business Rule Validation checks whether data follows rules defined by the business or organization.

Unlike simple validation such as checking whether a value is a number, business rule validation checks whether **different values make sense together according to a real-world business rule**.

For example:

* Salary cannot be negative.
* An employee's joining date cannot be before their date of birth.
* A student cannot have attendance above 100%.
* An order quantity must be greater than 0.
* An employee marked as `Inactive` should not have an active project.

## Simple Example

Suppose a student must have:

```text
Percentage between 0 and 100
Attendance between 0 and 100
```

If we have:

```text
Percentage = 85
Attendance = 90
```

the record is valid.

But:

```text
Percentage = 120
Attendance = 90
```

is invalid because percentage cannot be greater than 100.

## Real-World Example

In an online shopping system:

```text
Order_Quantity > 0
```

An order with:

```text
Quantity = 3 → Valid
Quantity = 0 → Invalid
Quantity = -2 → Invalid
```

because a customer cannot order zero or negative items.

## Business Example

Suppose an HR system has this business rule:

```text
Date of Joining >= Date of Birth
```

An employee cannot join a company before they were born.

Therefore:

```text
Date of Birth = 1995-05-10
Date of Joining = 2020-06-15
→ Valid
```

But:

```text
Date of Birth = 1995-05-10
Date of Joining = 1990-06-15
→ Invalid
```

## AI/ML Use Case

Business rule validation can be performed before machine learning training to remove logically incorrect records.

For example, in an employee salary prediction dataset:

```text
Experience = 10 years
Age = 5 years
```

is logically impossible.

Even though both values are numerical and within some numerical ranges, the record violates a business rule.



````python

## Python Example

import pandas as pd

df = pd.DataFrame({
    "Employee": ["Alice", "Bob", "Charlie"],
    "Age": [30, 25, 5],
    "Experience": [5, 8, 10]
})

# Business rule:
# Experience cannot be greater than Age
df["Rule_Valid"] = df["Experience"] <= df["Age"]

print(df)
````


## How the Output Comes

The important code is:

```python id="ujp5cu"
df["Experience"] <= df["Age"]
```

Python compares the two columns **row by row**.

For Alice:

```text
Experience = 5
Age = 30

5 <= 30
→ True
```

For Bob:

```text
Experience = 8
Age = 25

8 <= 25
→ True
```

For Charlie:

```text
Experience = 10
Age = 5

10 <= 5
→ False
```

Therefore:

```text id="qj3m2b"
Alice    → True
Bob      → True
Charlie  → False
```

Charlie fails the business rule because the employee's experience is greater than their age.

## Important Difference

Range Validation checks a value against a numerical limit:

```python id="8ov7ct"
Age >= 0
```

Business Rule Validation can compare **multiple columns** and check whether they make logical sense together:

```python id="rq6t9g"
Experience <= Age
```

So business rule validation is more about **business logic and relationships between data values**.

## Key Point

Business Rule Validation ensures that data follows the **real-world rules of the business or domain**, not just basic data-type or range rules

# Referential Validation

Referential Validation checks whether a value in one dataset or table correctly refers to an existing value in another dataset or table.

It is commonly used when two tables are connected through a common field such as:

* Customer ID
* Employee ID
* Product ID
* Department ID
* Order ID

The main purpose is to make sure that a reference actually exists.

## Simple Example

Suppose we have a Customer table:

```text id="t9h7p5"
Customer_ID
101
102
103
```

And an Orders table:

```text id="d7t1qp"
Order_ID  Customer_ID
1         101
2         102
3         105
```

Customer `105` does not exist in the Customer table.

Therefore:

```text id="7m0y2k"
101 → Valid
102 → Valid
105 → Invalid
```

## Real-World Example

An online shopping system may have a `Customer` table and an `Orders` table.

If an order contains:

```text id="l2e4zq"
Customer_ID = 101
```

then Customer `101` should exist in the Customer table.

If Customer `101` does not exist, the order contains an invalid customer reference.

## Business Example

Suppose a company has an Employee table:

```text id="v6q8mt"
Employee_ID
E001
E002
E003
```

The Payroll table contains:

```text id="1y8h6v"
Employee_ID  Salary
E001         50000
E002         60000
E005         45000
```

`E005` does not exist in the Employee table, so the payroll record for `E005` fails referential validation.

## AI/ML Use Case

Referential validation is useful when machine learning data is collected from multiple tables.

For example, customer transactions may be joined with customer information using `Customer_ID`.

If some transaction records contain customer IDs that do not exist in the customer table, the joined dataset may contain missing or incorrect information.

Validating the references before joining helps maintain data quality.



````python
## Python Example

import pandas as pd

customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Customer_Name": ["Alice", "Bob", "Charlie"]
})

orders = pd.DataFrame({
    "Order_ID": [1, 2, 3],
    "Customer_ID": [101, 102, 105]
})

# Check whether each order's Customer_ID exists
orders["Customer_Valid"] = orders["Customer_ID"].isin(customers["Customer_ID"])

print(orders)

````



## How the Output Comes

The important code is:

```python id="q4x1jz"
orders["Customer_ID"].isin(customers["Customer_ID"])
```

First, Python gets the valid Customer IDs from the customer table:

```text id="m2j7d8"
101
102
103
```

Then `.isin()` checks each `Customer_ID` from the orders table against that list.

For Order 1:

```text id="j6k3h2"
Customer_ID = 101

Is 101 present in [101, 102, 103]?
→ Yes
→ True
```

For Order 2:

```text id="g7q2s9"
Customer_ID = 102

Is 102 present in [101, 102, 103]?
→ Yes
→ True
```

For Order 3:

```text id="q8v5m1"
Customer_ID = 105

Is 105 present in [101, 102, 103]?
→ No
→ False
```

Therefore:

```text id="p4z7s0"
101 → True
102 → True
105 → False
```

`True` means the reference exists in the other table.

`False` means the reference does not exist and the record needs to be investigated.

## Key Point

Referential Validation checks:

**"Does this ID or reference actually exist in the related dataset or table?"**

For example:

```python id="c4x8qn"
orders["Customer_ID"].isin(customers["Customer_ID"])
```

means:

**"Check whether every Customer_ID in the Orders table exists in the Customers table."**

# Date Validation

Date Validation checks whether date values are valid, correctly formatted, and logically consistent with other dates.

Examples of date validation rules:

* Date should be a valid date.
* Date should not be in an invalid format.
* Date of joining cannot be before date of birth.
* Order date cannot be before customer registration date.
* Delivery date cannot be before order date.
* A future date may not be allowed for certain fields.

## Simple Example

Suppose an employee has:

```text
Date of Birth = 1995-05-10
Date of Joining = 2020-06-15
```

This is valid because the employee joined after their birth.

But:

```text
Date of Birth = 1995-05-10
Date of Joining = 1990-06-15
```

is invalid because the joining date is before the birth date.

## Real-World Example

In an online shopping system:

```text
Order Date    = 2026-08-01
Delivery Date = 2026-08-05
```

is valid.

But:

```text
Order Date    = 2026-08-05
Delivery Date = 2026-08-01
```

is invalid because an order cannot be delivered before it was placed.

## Business Example

An HR system can use the rule:

```text
Date of Joining >= Date of Birth
```

This prevents logically impossible employee records from entering the system.

## AI/ML Use Case

Date validation is important before creating machine learning features from dates.

For example, an incorrect date can produce an incorrect employee age, customer tenure, order duration, or time difference. This can negatively affect model training.




````python
## Python Example

import pandas as pd

df = pd.DataFrame({
    "Employee": ["Alice", "Bob", "Charlie"],
    "Date_of_Birth": ["1995-05-10", "2000-03-15", "1998-07-20"],
    "Date_of_Joining": ["2020-06-15", "1999-01-10", "2022-08-01"]
})

# Convert text values into datetime values
df["Date_of_Birth"] = pd.to_datetime(df["Date_of_Birth"])
df["Date_of_Joining"] = pd.to_datetime(df["Date_of_Joining"])

# Business rule:
# Date of Joining cannot be before Date of Birth
df["Date_Valid"] = df["Date_of_Joining"] >= df["Date_of_Birth"]

print(df)


````

## How the Output Comes

First, this code:

```python id="b2c5m8"
pd.to_datetime(df["Date_of_Birth"])
```

converts the date values from text into actual Pandas datetime values.

The same happens for:

```python id="t7k3p9"
pd.to_datetime(df["Date_of_Joining"])
```

Then Python applies the validation rule:

```python id="z4n6x2"
df["Date_of_Joining"] >= df["Date_of_Birth"]
```

Python compares the dates **row by row**.

For Alice:

```text id="h2p5q7"
Date of Joining = 2020-06-15
Date of Birth   = 1995-05-10

2020-06-15 >= 1995-05-10
→ True
```

Alice's joining date is after her birth date, so the record is valid.

For Bob:

```text id="w8c1r4"
Date of Joining = 1999-01-10
Date of Birth   = 2000-03-15

1999-01-10 >= 2000-03-15
→ False
```

Bob's joining date is before his birth date, so the record is invalid.

For Charlie:

```text id="p6y3t9"
Date of Joining = 2022-08-01
Date of Birth   = 1998-07-20

2022-08-01 >= 1998-07-20
→ True
```

Charlie passes the validation.

Therefore:

```text id="c7m4v2"
Alice    → True
Bob      → False
Charlie  → True
```

`True` means the date relationship is valid.

`False` means the date relationship violates the validation rule.

## Another Important Date Check

We can also check whether a date can be converted correctly:

```python id="j3k8s5"
pd.to_datetime("2026-08-24")
```

If the value is a valid date, Pandas converts it into a datetime value.

For invalid date text, we can use:

```python id="r5n2x7"
pd.to_datetime("invalid-date", errors="coerce")
```

The result becomes:

```text
NaT
```

`NaT` means **Not a Time**, which indicates that Pandas could not interpret the value as a valid date.

## Key Point

Date Validation checks both:

**1. Whether the date is valid.**

**2. Whether the date follows logical date relationships.**

For example:

```python id="m8q4z1"
df["Date_of_Joining"] >= df["Date_of_Birth"]
```

means:

**"Check whether every employee's joining date is on or after their date of birth."**

# Constraint Validation

Constraint Validation checks whether data follows predefined restrictions or conditions that must always be satisfied.

Constraints are rules that protect data from being incorrect, incomplete, or inconsistent.

Common constraints include:

* A value cannot be null.
* A value must be unique.
* A numerical value must be within a specific range.
* A value must belong to an accepted category.
* A value must satisfy a relationship with another value.

## Simple Example

Suppose every employee must have a unique Employee ID.

```text id="j6c3z9"
Employee_ID
E001
E002
E003
```

This is valid because every ID is unique.

But:

```text id="q7p1x4"
Employee_ID
E001
E002
E001
```

## Python Example

Suppose each Employee ID must be unique and every employee must have a name.



````python

import pandas as pd

df = pd.DataFrame({
    "Employee_ID": ["E001", "E002", "E001", "E004"],
    "Name": ["Alice", "Bob", None, "David"]
})

# Constraint 1: Employee_ID must be unique
df["ID_Valid"] = ~df["Employee_ID"].duplicated()

# Constraint 2: Name cannot be null
df["Name_Valid"] = df["Name"].notna()

# Overall validation
df["Record_Valid"] = df["ID_Valid"] & df["Name_Valid"]

print(df)
````

## How the Output Comes

### Step 1 – Checking Employee ID

The code:

```python id="b8r2n6"
df["Employee_ID"].duplicated()
```

checks whether an Employee ID has already appeared before.

The values are:

```text id="q3m7w1"
E001 → First occurrence → False
E002 → First occurrence → False
E001 → Already appeared → True
E004 → First occurrence → False
```

But we use `~` before it:

```python id="x5k9p2"
~df["Employee_ID"].duplicated()
```

`~` means **reverse the Boolean value**.

So:

```text id="s6d1y8"
False → True
False → True
True  → False
False → True
```

Therefore:

```text id="f1r4c7"
E001 → True
E002 → True
E001 → False
E004 → True
```

The second `E001` fails the uniqueness constraint.

### Step 2 – Checking Name

The code:

```python id="a7q3m9"
df["Name"].notna()
```

checks whether the Name is present.

```text id="u2v5k8"
Alice → Present → True
Bob   → Present → True
None  → Missing  → False
David → Present → True
```

Therefore:

```text id="r6t1p4"
Alice → True
Bob   → True
None  → False
David → True
```

### Step 3 – Checking Both Constraints

The code:

```python id="n8c4x2"
df["ID_Valid"] & df["Name_Valid"]
```

uses `&`, which means **AND**.

Both conditions must be `True`.

For Alice:

```text id="y5m2q7"
ID_Valid   = True
Name_Valid = True

True & True
→ True
```

For Bob:

```text id="k3r8v1"
True & True
→ True
```

For the third record:

```text id="d9w4p6"
ID_Valid   = False
Name_Valid = False

False & False
→ False
```

For David:

```text id="c2x7m5"
True & True
→ True
```

Therefore:

```text id="p8n3z6"
Alice → True
Bob   → True
E001  → False
David → True
```

`True` means the record satisfies all the checked constraints.

`False` means the record violates at least one constraint.

## Key Point

Constraint Validation ensures that important predefined rules are always satisfied.

In this example:

```python id="q6t2w9"
df["Record_Valid"] = df["ID_Valid"] & df["Name_Valid"]
```

means:

**"A record is valid only when its Employee ID is unique AND its Name is not missing."**
