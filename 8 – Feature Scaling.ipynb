# 1. Why Feature Scaling?

Feature Scaling is the process of transforming numerical features so that they are represented on a similar scale.

Different features in a dataset can have very different ranges. For example, `Age` may range from 18 to 70, while `Salary` may range from ₹20,000 to ₹2,00,000.


Here, `Age` contains values between 20 and 40, while `Salary` contains values between 20,000 and 1,00,000.

Because the numerical ranges are very different, some Machine Learning algorithms may give more influence to `Salary` simply because its values are much larger.

## Why Do We Need Feature Scaling?

Feature scaling helps to:

- Bring numerical features to comparable scales.
- Prevent large-valued features from dominating smaller-valued features.
- Improve distance calculations.
- Help optimization algorithms converge faster.
- Improve training stability for some Machine Learning algorithms.

## Real-World Example

Suppose a customer dataset contains:

    Age       → 18 to 70
    Income    → ₹20,000 to ₹2,00,000

If we use K-Nearest Neighbors (KNN) to find similar customers, the large income values can have a much greater effect on the distance calculation than age.

Feature scaling makes these features more comparable.

## Business Example

A bank wants to classify customers using:

- Age
- Annual Income
- Credit Score

Income may be measured in lakhs, while age ranges from 18 to 70 and credit score ranges from 300 to 900.

For scale-sensitive algorithms, scaling can prevent income from having an unnecessarily large numerical influence.

## AI/ML Use Case

Feature scaling is commonly important for algorithms such as:

- K-Nearest Neighbors (KNN)
- K-Means Clustering
- Support Vector Machine (SVM)
- Logistic Regression
- Linear Regression
- Neural Networks
- Principal Component Analysis (PCA)

Tree-based algorithms such as Decision Trees and Random Forests generally do not require feature scaling because they make decisions using feature thresholds.

## Important Point

Feature scaling does not remove the original information from a feature. It changes the numerical representation of the values so that different features can be represented on comparable scales.

For example:

    Before Scaling:

    Age       Salary
    20        20000
    30        50000
    40        100000

After scaling, the values might look like:

    Age       Salary
    0.0       0.0
    0.5       0.375
    1.0       1.0

The actual values have changed, but the information and relative ordering are still represented.

## Key Idea

Feature Scaling = Transforming numerical features to comparable scales so that scale-sensitive Machine Learning algorithms can work more effectively.

```python
## Simple Example

##Consider the following dataset:
import pandas as pd

df = pd.DataFrame({
"Age": [20, 30, 40],
"Salary": [20000, 50000, 100000]
    })

print(df)
```

# 2. Scale-Sensitive Algorithms

Scale-sensitive algorithms are Machine Learning algorithms whose performance or results can be affected by the numerical scale of input features.

If one feature has values much larger than another feature, the larger-valued feature may have more influence on the algorithm.

## Simple Example

Consider two features:

    Age       Salary
    20        20000
    30        50000
    40        100000

If an algorithm calculates the distance between customers, the difference in `Salary` is much larger than the difference in `Age`.

For example:

    Difference in Age = 10
    Difference in Salary = 30000

The salary values can dominate the distance calculation.

After scaling, both features can contribute more fairly.

## Algorithms Sensitive to Feature Scaling

Common scale-sensitive algorithms include:

- K-Nearest Neighbors (KNN)
- K-Means Clustering
- Support Vector Machine (SVM)
- Logistic Regression
- Linear Regression
- Neural Networks
- Principal Component Analysis (PCA)
- Gradient Descent-based algorithms

## Why Are They Scale-Sensitive?

Different algorithms use mathematical calculations such as:

- Distance
- Dot product
- Variance
- Gradients
- Magnitude of feature values

When features have very different ranges, these calculations can be dominated by features with larger numerical values.

## Real-World Example

Suppose an e-commerce company wants to group customers using K-Means.

The dataset contains:

    Age       → 18 to 70
    Income    → ₹20,000 to ₹2,00,000
    Spending  → ₹1,000 to ₹50,000

K-Means uses distance to determine which customers are similar.

Without scaling, `Income` can have a much larger influence because its numerical values are much larger.

Scaling the features helps K-Means compare customers more appropriately.

## Business Example

A bank uses KNN to classify loan applicants using:

- Age
- Income
- Credit Score

If the features are not scaled, income can dominate the distance calculation because its values are much larger than age or credit score.

Scaling allows the algorithm to consider all features on a more comparable numerical scale.

## AI/ML Use Case

Feature scaling is especially important when training models where distance, magnitude, variance, or gradient calculations are involved.

For example, when using KNN:

    Distance = sqrt((Age difference)^2 + (Income difference)^2)

If income values are very large, the income term can dominate the calculation.

After scaling:

    Distance = sqrt((Scaled Age difference)^2 + (Scaled Income difference)^2)

Now both features can contribute more appropriately.


The original features had very different ranges, but after standardization they are represented on a comparable scale.

## Key Idea

Scale-sensitive algorithms can be affected by differences in feature magnitude.

Therefore, feature scaling is usually important before using algorithms such as KNN, K-Means, SVM, PCA, and many gradient-based models.

```python

## Python Example

import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
        "Age": [20, 30, 40],
        "Salary": [20000, 50000, 100000]
    })

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

print(scaled_data)
```

# 3. Scale-Insensitive Algorithms

Scale-insensitive algorithms are Machine Learning algorithms whose results are generally not affected significantly by the numerical scale of input features.

These algorithms do not depend heavily on distance, feature magnitude, or the numerical range of features.

## Simple Example

Consider the following dataset:

    Age       Salary
    20        20000
    30        50000
    40        100000

A Decision Tree can split the data using conditions such as:

    Salary <= 50000

The tree only needs to determine where to split the feature. It does not require `Age` and `Salary` to be on the same scale.

If `Salary` is scaled, the split value will also change, but the relative ordering of the observations remains the same.

## Common Scale-Insensitive Algorithms

Common examples include:

- Decision Tree
- Random Forest
- Extra Trees
- Gradient Boosting Trees
- XGBoost
- LightGBM
- CatBoost

These algorithms are generally not sensitive to feature scaling.

## Why Are They Scale-Insensitive?

Tree-based algorithms make decisions using conditions such as:

    Age <= 30
    Salary <= 50000

The algorithm mainly considers the ordering of values and determines suitable split points.

For example:

    Salary:
    20000
    50000
    100000

After scaling:

    0.0
    0.375
    1.0

The numerical values changed, but their ordering remains:

    20000 < 50000 < 100000

Therefore, the tree can still identify similar splits.

## Real-World Example

Suppose a company wants to predict whether a customer will purchase a product using:

- Age
- Income
- Website Visits

A Decision Tree can create rules such as:

    Income <= 50000
    Website Visits <= 5
    Age <= 30

The different numerical ranges do not usually create a problem for the tree.

## Business Example

A bank uses a Random Forest model to predict loan approval using:

- Age
- Annual Income
- Credit Score
- Loan Amount

The income and loan amount may contain large numbers, while age has smaller values.

Random Forest does not generally require these features to be scaled before training.

## AI/ML Use Case

Tree-based algorithms are often useful when:

- Features have different numerical ranges.
- Features contain mixed numerical scales.
- Scaling would add unnecessary preprocessing.
- The model is based on decision rules and feature thresholds.


Output:

    Model trained successfully

Scaling is not required here for the Decision Tree to work effectively.

## Important Point

Scale-insensitive does not mean that scaling is always forbidden.

You can still scale the data, but for tree-based algorithms it usually does not provide a meaningful advantage.

## Key Idea

Scale-insensitive algorithms generally do not require feature scaling because their calculations are not strongly affected by differences in feature magnitude.

**Decision Trees and tree-based models are the most common examples of scale-insensitive algorithms.**

```python

## Python Example

import pandas as pd
from sklearn.tree import DecisionTreeClassifier

df = pd.DataFrame({
        "Age": [20, 25, 30, 35, 40, 45],
        "Salary": [20000, 30000, 40000, 60000, 80000, 100000],
        "Purchased": [0, 0, 0, 1, 1, 1]
    })

X = df[["Age", "Salary"]]
y = df["Purchased"]

model = DecisionTreeClassifier(random_state=42)

model.fit(X, y)

print("Model trained successfully")
```

# 4. Normalization

Normalization is a feature scaling technique that transforms numerical values into a specific range, commonly between 0 and 1.

The main purpose of normalization is to make features comparable when they have different numerical ranges.

## Simple Example

Suppose we have the following values:

    Age
    20
    30
    40
    50

The values range from 20 to 50.

After normalization, they can be transformed into:

    0.00
    0.33
    0.67
    1.00

Now the values are represented between 0 and 1.

## Formula

The commonly used Min-Max normalization formula is:

    X_scaled = (X - X_min) / (X_max - X_min)

Where:

    X       = Original value
    X_min   = Minimum value
    X_max   = Maximum value
    X_scaled = Normalized value

## Calculation Example

Suppose the values are:

    Age = [20, 30, 40, 50]

For `Age = 30`:

    X = 30
    X_min = 20
    X_max = 50

Therefore:

    X_scaled = (30 - 20) / (50 - 20)
             = 10 / 30
             = 0.33

So, the normalized value of 30 is approximately `0.33`.

## Real-World Example

Suppose a Machine Learning model uses:

    Age       → 18 to 70
    Income    → 20000 to 200000
    Experience → 0 to 30

These features have different ranges.

Normalization transforms them into a common range such as:

    0 to 1

This makes the feature values more comparable.

## Business Example

An e-commerce company wants to group customers based on:

- Number of purchases
- Total spending
- Website visits

The ranges may be very different:

    Purchases       → 1 to 50
    Total Spending  → ₹1,000 to ₹1,00,000
    Website Visits  → 1 to 500

Normalization can transform all these features into a similar range before using a scale-sensitive algorithm such as K-Means.

## AI/ML Use Case

Normalization is commonly useful for:

- K-Nearest Neighbors
- K-Means Clustering
- Neural Networks
- Support Vector Machines
- Distance-based algorithms

It is particularly useful when features have known or meaningful minimum and maximum values and you want the transformed values to lie within a fixed range.

## Important Point

Normalization and Min-Max Scaling are often used interchangeably in Machine Learning discussions.

In practical Machine Learning, when people say "normalize the features to 0–1", they usually mean applying Min-Max Scaling.

## Limitation

Normalization is sensitive to outliers.

For example:

    Salary = [20000, 30000, 40000, 50000, 1000000]

The value `1000000` becomes the maximum value.

Because Min-Max scaling uses the minimum and maximum values, most of the other salaries can become compressed close to 0.

Therefore, normalization may not be the best choice when strong outliers are present.

## Key Idea

**Normalization transforms numerical features into a common range, commonly 0 to 1, making them more comparable for scale-sensitive Machine Learning algorithms.**

```python

## Python Example

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.DataFrame({
"Age": [20, 30, 40, 50],
"Salary": [20000, 50000, 80000, 100000]
    })

scaler = MinMaxScaler()

normalized_data = scaler.fit_transform(df)

print(normalized_data)


#Both `Age` and `Salary` have now been transformed to values between 0 and 1.
```

# 5. Standardization

Standardization is a feature scaling technique that transforms numerical data so that it has a mean of approximately 0 and a standard deviation of approximately 1.

Unlike Min-Max Scaling, standardization does not force values into a fixed range such as 0 to 1.

## Simple Example

Suppose we have:

    Age
    20
    30
    40
    50

After standardization, the values may become approximately:

    -1.34
    -0.45
     0.45
     1.34

The values are now centered around 0.

## Formula

The standardization formula is:

    X_scaled = (X - Mean) / Standard Deviation

Where:

    X                  = Original value
    Mean               = Mean of the feature
    Standard Deviation = Standard deviation of the feature
    X_scaled           = Standardized value

## How Does It Work?

Suppose:

    Age = [20, 30, 40, 50]

The mean is:

    Mean = (20 + 30 + 40 + 50) / 4
         = 35

For the value `20`:

    X_scaled = (20 - 35) / Standard Deviation

The result will be a negative value because 20 is below the mean.

For the value `50`, the result will be positive because 50 is above the mean.

Therefore:

    Below Mean → Negative value
    At Mean    → Approximately 0
    Above Mean → Positive value

## Real-World Example

Suppose a dataset contains:

    Age
    Income
    Credit Score

These features have very different numerical ranges.

Standardization transforms each feature independently so that its values are centered around 0 with a standard deviation of 1.

## Business Example

A bank uses Logistic Regression to predict whether a customer will default on a loan.

The model uses:

    Age

- Principal Component Analysis

## Standardization vs Normalization

Normalization commonly transforms values into a fixed range such as 0 to 1.

Standardization transforms values based on the mean and standard deviation.

    Normalization:
    Usually → 0 to 1

    Standardization:
    Mean → 0
    Standard Deviation → 1

Standardized values can be less than 0 or greater than 1.

## Important Point

Standardization does not guarantee that all values will fall between 0 and 1.

For example:

    -2.0
    -1.0
     0.0
     1.0
     2.0

These are valid standardized values.

## Limitation

Standardization is affected by outliers because the mean and standard deviation can change significantly when extreme values are present.

When a dataset contains strong outliers, `RobustScaler` may be a better option.

## Key Idea

**Standardization transforms features so that they have approximately mean 0 and standard deviation 1, making them suitable for many scale-sensitive Machine Learning algorithms.**

```python


## Python Example

import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
"Age": [20, 30, 40, 50]
    })

scaler = StandardScaler()

standardized_data = scaler.fit_transform(df)

print(standardized_data)

#The mean of the transformed values is approximately 0, and the standard deviation is approximately 1.
```

# 6. Min-Max Scaling

Min-Max Scaling is a feature scaling technique that transforms numerical values into a specified range, most commonly 0 to 1.

It is a type of normalization and is useful when we want all features to have a fixed and comparable range.

## Simple Example

Suppose we have the following values:

    Age
    20
    30
    40
    50

The minimum value is 20 and the maximum value is 50.

After Min-Max Scaling:

    20 → 0.00
    30 → 0.33
    40 → 0.67
    50 → 1.00

The smallest value becomes 0 and the largest value becomes 1.

## Formula

The Min-Max Scaling formula is:

    X_scaled = (X - X_min) / (X_max - X_min)

Where:

    X       = Original value
    X_min   = Minimum value
    X_max   = Maximum value
    X_scaled = Scaled value

## Calculation Example

Suppose:

    Age = [20, 30, 40, 50]

For `Age = 30`:

    X = 30
    X_min = 20
    X_max = 50

Therefore:

    X_scaled = (30 - 20) / (50 - 20)
             = 10 / 30
             = 0.33

So, `30` becomes approximately `0.33`.


## Understanding the Output

For `Age`:

    Minimum Age = 20 → 0
    Maximum Age = 50 → 1

For `Salary`:

    Minimum Salary = 20000 → 0
    Maximum Salary = 100000 → 1

The values in between are converted proportionally.

## Real-World Example

Suppose a dataset contains:

    Age          → 18 to 70
    Income       → ₹20,000 to ₹2,00,000
    Experience   → 0 to 30 years

These features have very different ranges.

Min-Max Scaling converts them into a common range:

    0 to 1

This makes them more comparable for algorithms that are sensitive to feature scale.

## Business Example

An e-commerce company wants to group customers using K-Means.

The dataset contains:

    Purchases       → 1 to 50
    Total Spending  → ₹1,000 to ₹1,00,000
    Website Visits  → 1 to 500

Without scaling, total spending can dominate the distance calculation.

Min-Max Scaling transforms these features into a common range so that K-Means can use them more appropriately.

## AI/ML Use Case

Min-Max Scaling is commonly useful for:

- K-Nearest Neighbors
- K-Means Clustering
- Neural Networks
- Support Vector Machines
- Distance-based algorithms

It is also useful when a model performs better when input values are within a small fixed range.

## Important Point

Min-Max Scaling does not change the relative ordering of the values.

For example:

    Original:
    20 < 30 < 40 < 50

After scaling:

    0 < 0.33 < 0.67 < 1

The order remains the same.

## Effect of Outliers

Min-Max Scaling is sensitive to outliers because it uses the minimum and maximum values.

For example:

    Salary = [20000, 30000, 40000, 50000, 1000000]

Here, `1000000` becomes the maximum.

As a result, the smaller salary values may become compressed close to 0 after scaling.

Therefore, Min-Max Scaling should be used carefully when the dataset contains extreme outliers.

## Key Idea

**Min-Max Scaling transforms numerical features into a fixed range, commonly 0 to 1, using the minimum and maximum values of each feature.**

```python

## Python Example

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.DataFrame({
        "Age": [20, 30, 40, 50],
        "Salary": [20000, 50000, 80000, 100000]
})

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(df)

print(scaled_data)


#Each feature has been transformed independently into the range 0 to 1.
```

# 7. StandardScaler

`StandardScaler` is a preprocessing tool from Scikit-learn used to standardize numerical features.

It transforms each feature so that its mean becomes approximately 0 and its standard deviation becomes approximately 1.

## Simple Example

Suppose we have:

    Age
    20
    30
    40
    50

We can use `StandardScaler` to transform these values into standardized values.

## How `StandardScaler` Works

`StandardScaler` uses the standardization formula:

    X_scaled = (X - Mean) / Standard Deviation

For each feature, it calculates:

    1. Mean
    2. Standard Deviation
    3. Standardized value for every observation

For example, if the mean age is 35, a value below 35 will produce a negative standardized value, while a value above 35 will produce a positive standardized value.

## `fit()`

`fit()` calculates the statistics required for scaling.

    scaler.fit(df)

It learns:

    Mean
    Standard Deviation

from the training data.

It does not transform the data yet.

## `transform()`

`transform()` uses the statistics learned during `fit()` to transform the data.

    scaler.transform(df)

## `fit_transform()`

`fit_transform()` performs both operations together:

    scaler.fit_transform(df)

It first learns the mean and standard deviation and then transforms the data.

## Real-World Example

Suppose a customer dataset contains:

    Age
    Income
    Credit Score

The features have different numerical ranges.

Using `StandardScaler` transforms each feature independently so that the features are centered around 0 with a standard deviation of approximately 1.

## Business Example

A bank wants to build a Logistic Regression model for loan default prediction.

The input features are:

    Age
    Income
    Loan Amount
    Credit Score

Since these features have different scales, `StandardScaler` can be used before training the model.

## AI/ML Use Case

`StandardScaler` is commonly used with:

- Logistic Regression
- Linear Regression
- Support Vector Machine
- K-Nearest Neighbors
- Principal Component Analysis
- Neural Networks
- Gradient Descent-based algorithms

## Important Point

`StandardScaler` does not restrict values to 0 and 1.

The transformed values can be:

    -2
    -1
     0
     1
     2

A value of approximately 0 means the observation is close to the feature mean.

A positive value means the observation is above the mean.

A negative value means the observation is below the mean.

## Outlier Effect

`StandardScaler` is sensitive to outliers because it uses the mean and standard deviation.

For example:

    Salary = [20000, 30000, 40000, 50000, 1000000]

The extreme value `1000000` can significantly increase the mean and standard deviation.

This can affect the standardized values of the other observations.

When strong outliers are present, `RobustScaler` may be a better choice.

## Key Idea

**`StandardScaler` standardizes numerical features by subtracting the mean and dividing by the standard deviation, resulting in features with approximately mean 0 and standard deviation 1.**

```python

## Python Example

import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
        "Age": [20, 30, 40, 50],
        "Salary": [20000, 50000, 80000, 100000]
    })

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

print(scaled_data)


#The original `Age` and `Salary` values had very different scales.

#After using `StandardScaler`, both features are represented around a mean of 0.
```

# 8. MinMaxScaler

`MinMaxScaler` is a Scikit-learn preprocessing tool used to scale numerical features to a specified range.

By default, `MinMaxScaler` transforms values into the range 0 to 1.

## Simple Example

Suppose we have:

    Age
    20
    30
    40
    50

The minimum value is 20 and the maximum value is 50.

After applying `MinMaxScaler`:

    20 → 0.00
    30 → 0.33
    40 → 0.67
    50 → 1.00


## How `MinMaxScaler` Works

`MinMaxScaler` uses the Min-Max Scaling formula:

    X_scaled = (X - X_min) / (X_max - X_min)

For each feature, it finds:

    Minimum value
    Maximum value

It then transforms every value based on those two values.

## `fit()`

`fit()` learns the minimum and maximum values from the data.

    scaler.fit(df)

For example:

    Age:
    Minimum = 20
    Maximum = 50

It stores these values for later transformation.

## `transform()`

`transform()` uses the learned minimum and maximum values to scale the data.

    scaler.transform(df)

## `fit_transform()`

`fit_transform()` performs both operations together.

    scaler.fit_transform(df)

It first learns the minimum and maximum values and then scales the data.

## Real-World Example

Suppose a dataset contains:

    Age          → 18 to 70
    Income       → ₹20,000 to ₹2,00,000
    Experience   → 0 to 30 years

The features have different ranges.

`MinMaxScaler` converts each feature independently into a common range of 0 to 1.

## Business Example

An e-commerce company wants to use K-Means to group customers.

The features include:

    Number of Purchases
    Total Spending
    Website Visits

Their ranges may be very different.

`MinMaxScaler` can transform all of them into the range 0 to 1 before applying K-Means.

This helps prevent a feature with larger numerical values from dominating the distance calculation.

## AI/ML Use Case

`MinMaxScaler` is commonly useful for:

- K-Nearest Neighbors
- K-Means Clustering
- Neural Networks
- Support Vector Machines
- Distance-based algorithms

## Changing the Scaling Range

By default:

    feature_range = (0, 1)

But we can specify another range.

For example:

    scaler = MinMaxScaler(feature_range=(-1, 1))

This transforms the features into the range -1 to 1.

## Important Point

`MinMaxScaler` preserves the relative ordering of values.

For example:

    Original:
    20 < 30 < 40 < 50

After scaling:

    0 < 0.33 < 0.67 < 1

The values change, but their order remains the same.

## Outlier Effect

`MinMaxScaler` is sensitive to outliers because it depends on the minimum and maximum values.

For example:

    Salary = [20000, 30000, 40000, 50000, 1000000]

The value `1000000` becomes the maximum.

As a result, the other values can become compressed into a small range close to 0.

When strong outliers are present, `RobustScaler` may be more suitable.

## Key Idea

**`MinMaxScaler` transforms numerical features into a specified range, usually 0 to 1, using the minimum and maximum values of each feature.**

```python

## Python Example

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.DataFrame({
        "Age": [20, 30, 40, 50],
        "Salary": [20000, 50000, 80000, 100000]
})

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(df)

print(scaled_data)
```

# 9. RobustScaler

`RobustScaler` is a Scikit-learn preprocessing tool used to scale numerical features while reducing the influence of outliers.

Unlike `StandardScaler`, which uses the mean and standard deviation, `RobustScaler` uses the median and interquartile range (IQR).

## Why Do We Need RobustScaler?

Consider the following data:

    Salary
    20000
    30000
    40000
    50000
    1000000

Here, `1000000` is an extreme value.

Mean and standard deviation can be strongly affected by this outlier.

`RobustScaler` uses the median and IQR, which are less affected by extreme values.

## Simple Example

Suppose:

    Salary = [20000, 30000, 40000, 50000, 1000000]

The median is:

    Median = 40000

The median is not heavily affected by the extreme value `1000000`.

The IQR is calculated using:

    IQR = Q3 - Q1

Where:

    Q1 = First Quartile
    Q3 = Third Quartile

`RobustScaler` uses these values to scale the data.

## Formula

The basic formula is:

    X_scaled = (X - Median) / IQR

Where:

    X       = Original value
    Median  = Median of the feature
    IQR     = Interquartile Range
    X_scaled = Scaled value


## How `RobustScaler` Works

`RobustScaler` performs these main steps:

    1. Calculate the median.
    2. Calculate Q1.
    3. Calculate Q3.
    4. Calculate IQR = Q3 - Q1.
    5. Subtract the median from each value.
    6. Divide by the IQR.

This makes the scaling more resistant to outliers.

## Understanding Median

The median is the middle value when the data is arranged in order.

For example:

    [10, 20, 30, 40, 50]

Median:

    30

If we add an extreme value:

    [10, 20, 30, 40, 1000]

The median is still:

    30

This is why median is more resistant to extreme values than the mean.

## Understanding IQR

IQR represents the range of the middle 50% of the data.

It is calculated as:

    IQR = Q3 - Q1

For example:

    Q1 = 20
    Q3 = 50

Therefore:

    IQR = 50 - 20
        = 30

Because IQR focuses on the middle portion of the data, extreme values have less influence on it.

## Real-World Example

Consider employee salaries:

    25000
    30000
    35000
    40000
    45000
    50000
    500000

The value `500000` may be an executive salary and can act as an outlier.

Using `StandardScaler` or `MinMaxScaler` can cause the outlier to strongly affect the scaling.

`RobustScaler` is more appropriate when we want to reduce the influence of such extreme values.

## Business Example

A company analyzes customer spending:

    ₹1,000
    ₹2,000
    ₹3,000
    ₹4,000
    ₹5,000
    ₹1,00,000

A small number of customers may spend extremely large amounts.

If these outliers strongly affect the scaling, normal scaling techniques may compress the majority of customers into a small range.

`RobustScaler` can handle this situation better because it uses the median and IQR.

## AI/ML Use Case

`RobustScaler` can be useful for:

- K-Nearest Neighbors
- K-Means Clustering
- Support Vector Machines
- Logistic Regression
- Linear Regression
- Neural Networks

It is especially useful when numerical features contain significant outliers.

## Comparison with StandardScaler

`StandardScaler`:

    Uses Mean
    Uses Standard Deviation
    Sensitive to Outliers

`RobustScaler`:

    Uses Median
    Uses IQR
    Less Sensitive to Outliers

## Important Point

`RobustScaler` does not remove outliers.

It only reduces the influence of outliers when calculating the scaling parameters.

The outlier itself remains in the transformed dataset.

## Key Idea

**`RobustScaler` scales numerical features using the median and IQR, making it more resistant to outliers than StandardScaler and MinMaxScaler.**

```python

## Python Example

import pandas as pd
from sklearn.preprocessing import RobustScaler

df = pd.DataFrame({
        "Salary": [20000, 30000, 40000, 50000, 1000000]
    })

scaler = RobustScaler()

scaled_data = scaler.fit_transform(df)

print(scaled_data)

#Output will contain scaled values where the median value is centered around 0, while the extreme value remains large but has less influence on the scaling calculation.
```

# 10. MaxAbsScaler

`MaxAbsScaler` is a Scikit-learn preprocessing technique that scales each numerical feature using its maximum absolute value.

It transforms values so that they generally fall between `-1` and `1`.

Unlike `MinMaxScaler`, `MaxAbsScaler` does not shift the data toward a new minimum or maximum. It mainly changes the magnitude of the values.

## Simple Example

Suppose we have:

    Score
    -50
    -25
     0
     25
     50

The maximum absolute value is:

    50

After applying MaxAbsScaler:

    -50 → -1.0
    -25 → -0.5
      0 →  0.0
     25 →  0.5
     50 →  1.0

The values are now within the range `-1` to `1`.

## Formula

The formula is:

    X_scaled = X / max(|X|)

Where:

    X = Original value
    max(|X|) = Maximum absolute value in the feature
    X_scaled = Scaled value

## Calculation Example

Suppose:

    Score = [-50, -25, 0, 25, 50]

Maximum absolute value:

    max(|X|) = 50

For `X = 25`:

    X_scaled = 25 / 50
             = 0.5

For `X = -50`:

    X_scaled = -50 / 50
             = -1.0

## How `MaxAbsScaler` Works

For each feature, `MaxAbsScaler` finds the largest absolute value.

For example:

    Feature:
    [-50, -25, 0, 25, 50]

Maximum absolute value:

    50

Every value is divided by 50.

Therefore:

    -50 / 50 = -1
    -25 / 50 = -0.5
      0 / 50 = 0
     25 / 50 = 0.5
     50 / 50 = 1

## Important Property

`MaxAbsScaler` does not center the data around 0 by subtracting the mean.

It preserves:

- The sign of the values.
- Zero values.
- Sparsity of sparse datasets.

For example:

    Original:
    [-100, 0, 50]

After scaling:

    [-1, 0, 0.5]

The zero value remains zero.

## Real-World Example

Suppose a financial dataset contains changes in account balance:

    -5000
    -2000
     0
     3000
     5000

Negative values represent withdrawals or losses, while positive values represent deposits or gains.

`MaxAbsScaler` can scale these values while preserving their positive and negative signs.

## Business Example

A company analyzes monthly profit and loss:

    -₹50,000
    -₹20,000
     ₹0
     ₹30,000
     ₹50,000

Here, both negative and positive values have business meaning.

`MaxAbsScaler` can scale the values while preserving the direction:

    Loss → Negative
    No Change → 0
    Profit → Positive

## AI/ML Use Case

`MaxAbsScaler` can be useful when:

- Data contains both positive and negative values.
- Zero values are meaningful.
- The dataset is sparse.
- We want to preserve the sign of the values.
- Features have different magnitudes.

It can be useful with algorithms such as:

- Linear Models
- Logistic Regression
- Neural Networks
- Support Vector Machines

## Outlier Effect

`MaxAbsScaler` is sensitive to extreme values because it uses the maximum absolute value.

For example:

    Score = [-10, -5, 0, 5, 1000]

The maximum absolute value is:

    1000

After scaling:

    -10 → -0.01
    -5  → -0.005
     0  → 0
     5  → 0.005
     1000 → 1

Most of the normal values become compressed close to zero.

Therefore, `MaxAbsScaler` is not a good choice when strong extreme outliers are present.

## Comparison with MinMaxScaler

`MinMaxScaler`:

    Uses minimum and maximum values.
    Usually transforms data to 0 to 1.
    Can shift the data.
    Sensitive to outliers.

`MaxAbsScaler`:

    Uses maximum absolute value.
    Usually transforms data to -1 to 1.
    Does not shift the data.
    Preserves zero and sign.
    Sensitive to extreme values.

## Key Idea

**`MaxAbsScaler` scales each feature by its maximum absolute value, usually producing values between -1 and 1 while preserving the sign and zero values.**

```python

## Python Example

import pandas as pd
from sklearn.preprocessing import MaxAbsScaler

df = pd.DataFrame({
        "Score": [-50, -25, 0, 25, 50],
        "Balance": [-1000, -500, 0, 500, 1000]
    })

scaler = MaxAbsScaler()

    
scaled_data = scaler.fit_transform(df)

print(scaled_data)
#Each feature has been divided by its maximum absolute value.
```

# 11. Comparison of Scaling Techniques

Different scaling techniques transform numerical features in different ways. The correct technique depends on the data, the Machine Learning algorithm, and whether the dataset contains outliers.

## Scaling Techniques

The main scaling techniques covered in this notebook are:

    Normalization
    Standardization
    Min-Max Scaling
    StandardScaler
    MinMaxScaler
    RobustScaler
    MaxAbsScaler

## Comparison Table

| Technique | Main Idea | Output Range | Outlier Sensitivity | Best Used When |
|---|---|---|---|---|
| Normalization | Scale values to a common range | Usually 0 to 1 | High | Features need a fixed range |
| Standardization | Center using mean and standard deviation | No fixed range | High | Data is reasonably distributed |
| Min-Max Scaling | Use minimum and maximum values | Usually 0 to 1 | High | A fixed range is required |
| StandardScaler | Mean = 0, Standard Deviation = 1 | No fixed range | High | General-purpose standardization |
| MinMaxScaler | Scale using minimum and maximum | Usually 0 to 1 | High | Distance-based algorithms |
| RobustScaler | Use median and IQR | No fixed range | Low | Data contains outliers |
| MaxAbsScaler | Divide by maximum absolute value | Usually -1 to 1 | High | Sparse data or signed values |

## Normalization vs Standardization

Normalization generally scales values into a fixed range such as 0 to 1.

Standardization transforms the feature so that:

    Mean ≈ 0
    Standard Deviation ≈ 1

Example:

    Original:
    20, 30, 40, 50

Normalization:

    0.00, 0.33, 0.67, 1.00

Standardization:

    -1.34, -0.45, 0.45, 1.34

## MinMaxScaler vs StandardScaler

`MinMaxScaler`:

    Uses minimum and maximum.
    Usually produces values between 0 and 1.
    Sensitive to outliers.

`StandardScaler`:

    Uses mean and standard deviation.
    Does not have a fixed output range.
    Sensitive to outliers.

## StandardScaler vs RobustScaler

`StandardScaler` uses:

    Mean
    Standard Deviation

`RobustScaler` uses:

    Median
    IQR

Therefore:

    StandardScaler → Better when there are no strong outliers.
    RobustScaler   → Better when strong outliers are present.

## MinMaxScaler vs RobustScaler

`MinMaxScaler` depends on:

    Minimum
    Maximum

Therefore, an extreme value can significantly affect the scaling.

`RobustScaler` depends on:

    Median
    IQR

Therefore, extreme values have much less influence on the scaling parameters.

## MaxAbsScaler

`MaxAbsScaler` is useful when:

- Data contains positive and negative values.
- Zero values need to remain zero.
- Sparse data needs to be preserved.
- Features have different magnitudes.

However, it is sensitive to extreme values because it uses the maximum absolute value.

## Which Technique Should You Choose?

A simple decision guide is:

    No major outliers
        → StandardScaler

    Need values between 0 and 1
        → MinMaxScaler

    Strong outliers are present
        → RobustScaler

    Sparse data or signed values
        → MaxAbsScaler

    Need a fixed range for a scale-sensitive model
        → MinMaxScaler

## Real-World Example

Suppose a customer dataset contains:

    Age
    Income
    Spending
    Credit Score

If there are no significant outliers and the model is Logistic Regression, `StandardScaler` can be a good general-purpose choice.

If Income contains extreme values, `RobustScaler` may be more appropriate.

If the model requires features between 0 and 1, `MinMaxScaler` can be used.

## Business Example

An e-commerce company uses K-Means to group customers based on:

    Number of Purchases
    Total Spending
    Website Visits

Because K-Means uses distances, scaling is important.

If the features contain strong outliers in spending, `RobustScaler` may be preferable.

If the features have relatively normal ranges without extreme outliers, `MinMaxScaler` or `StandardScaler` can be considered.

## AI/ML Use Case

There is no single scaling technique that is always the best.

The choice depends on:

    Algorithm
    Feature distribution
    Outliers
    Required output range
    Presence of negative values
    Sparse data

## Important Point

Scaling should be learned from the training data only.

Correct approach:

    scaler.fit(X_train)
    X_train_scaled = scaler.transform(X_train)
    X_test_scaled = scaler.transform(X_test)

Do not fit the scaler separately on the test data.

The test data must be transformed using the same scaling parameters learned from the training data.

## Key Idea

**There is no universal best scaler. Choose the scaling technique based on the data characteristics, outliers, algorithm requirements, and desired output range.**

```python

## Simple Python Comparison

import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, MaxAbsScaler

df = pd.DataFrame({
        "Age": [20, 30, 40, 50, 60],
        "Salary": [20000, 40000, 60000, 80000, 100000]
    })

standard = StandardScaler()
minmax = MinMaxScaler()
robust = RobustScaler()
maxabs = MaxAbsScaler()

print("StandardScaler:")
print(standard.fit_transform(df))

print("\nMinMaxScaler:")
print(minmax.fit_transform(df))

print("\nRobustScaler:")
print(robust.fit_transform(df))

print("\nMaxAbsScaler:")
print(maxabs.fit_transform(df))

#Each scaler produces different numerical values because each technique uses a different mathematical method.
```

# 12. Effect of Outliers on Scaling

Outliers are unusually large or small values that are significantly different from most observations in a dataset.

Outliers can strongly affect some scaling techniques because those techniques use statistics such as the minimum, maximum, mean, or standard deviation.

## Simple Example

Consider the following salary data:

    Salary
    20000
    30000
    40000
    50000
    1000000

Here, `1000000` is an extreme value compared with the other salaries.

This outlier can strongly affect scaling.

## Effect on MinMaxScaler

`MinMaxScaler` uses:

    Minimum
    Maximum

For the data:

    [20000, 30000, 40000, 50000, 1000000]

The maximum is:

    1000000

Because `1000000` is extremely large, most other values become very close to 0 after scaling.

For example, approximately:

    20000  → 0.000
    30000  → 0.010
    40000  → 0.020
    50000  → 0.031
    1000000 → 1.000

The majority of the observations are compressed into a very small range.

## Effect on StandardScaler

`StandardScaler` uses:

    Mean
    Standard Deviation

An extreme outlier can significantly increase both the mean and standard deviation.

For example:

    Salary = [20000, 30000, 40000, 50000, 1000000]

The value `1000000` pulls the mean upward and increases the standard deviation.

As a result, the standardized values of the normal observations can become compressed closer together.

## Effect on RobustScaler

`RobustScaler` uses:

    Median
    IQR

The median and IQR are much less affected by extreme values.

Therefore, when a dataset contains strong outliers, `RobustScaler` usually preserves the relative spread of the majority of the observations better than MinMaxScaler or StandardScaler.

## Comparison

For data containing strong outliers:

    MinMaxScaler
        → Highly affected by minimum and maximum.

    StandardScaler
        → Affected by mean and standard deviation.

    RobustScaler
        → Less affected because it uses median and IQR.

    MaxAbsScaler
        → Affected by the maximum absolute value.


## Real-World Example

Suppose most employees in a company earn between:

    ₹20,000 and ₹80,000

but one executive earns:

    ₹10,00,000

If we apply Min-Max Scaling, the executive salary becomes the maximum value and can cause most employee salaries to be compressed close to 0.

If we use RobustScaler, the median and IQR are used, so the executive salary has much less influence on the scaling of normal employee salaries.

## Business Example

An e-commerce company analyzes customer spending:

    ₹1,000
    ₹2,000
    ₹3,000
    ₹4,000
    ₹5,000
    ₹1,00,000

The ₹1,00,000 customer is an extreme spender.

If the company uses a distance-based algorithm such as K-Means, inappropriate scaling can make the normal customers appear very close together.

Using `RobustScaler` can reduce the influence of the extreme spending value during scaling.

## AI/ML Use Case

Outlier-aware scaling is particularly important for algorithms such as:

    K-Nearest Neighbors
    K-Means
    Support Vector Machine
    Logistic Regression
    Neural Networks

These algorithms can be sensitive to feature magnitudes, so extreme values can affect model behavior.

## Important Point

`RobustScaler` does not remove outliers.

It only reduces their influence when calculating the scaling parameters.

If an outlier is actually an error, it may need to be corrected or removed using an appropriate outlier treatment method.

If the outlier is a valid observation, it should generally not be removed blindly.

## Practical Decision

A simple approach is:

    No significant outliers
        → StandardScaler or MinMaxScaler

    Strong outliers present
        → Consider RobustScaler

    Sparse data with meaningful zero values
        → Consider MaxAbsScaler

Always inspect the data before choosing a scaling technique.

## Key Idea

**Outliers can strongly affect MinMaxScaler, StandardScaler, and MaxAbsScaler because they depend on extreme values or statistics influenced by extreme values. RobustScaler is more resistant to outliers because it uses the median and IQR.**

```python

## Python Example

import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler

df = pd.DataFrame({
    "Salary": [20000, 30000, 40000, 50000, 1000000]
})

minmax = MinMaxScaler()
standard = StandardScaler()
robust = RobustScaler()

print("MinMaxScaler:")
print(minmax.fit_transform(df))

print("\nStandardScaler:")
print(standard.fit_transform(df))

print("\nRobustScaler:")
print(robust.fit_transform(df))

#The three scalers produce different results because they use different statistics.
```
