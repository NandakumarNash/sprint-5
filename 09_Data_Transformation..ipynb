# Why Transform Data?

Data Transformation means changing the values or distribution of data into a more suitable form for analysis and machine learning.

## Why Do We Transform Data?

Data may contain highly skewed values, extreme values, large differences in magnitude, or distributions that are not suitable for certain machine learning algorithms.

Transformation can help make the data more suitable for statistical analysis and machine learning.

## Main Reasons for Data Transformation

1. Reduce skewness
2. Reduce the influence of extreme values
3. Make the distribution more suitable for analysis
4. Improve the performance of some machine learning algorithms
5. Improve numerical stability
6. Make relationships between variables easier to model

## Simple Example

Suppose we have income data:

    income = [20000, 22000, 25000, 28000, 30000, 35000, 40000, 500000]

Most values are between ₹20,000 and ₹40,000, but one value is ₹5,00,000.

This creates a right-skewed distribution.

The large value can have a strong influence on some statistical and machine learning methods.

## Applying Log Transformation

    import numpy as np

    income = [20000, 22000, 25000, 28000, 30000, 35000, 40000, 500000]

    log_income = np.log(income)

    print(log_income)

The log transformation compresses large values.

For example:

    Original:  20,000  → Log:  9.90
    Original:  40,000  → Log: 10.60
    Original: 500,000  → Log: 13.12

Notice that the difference between small and large values becomes much smaller after transformation.



## Output Explanation

Before transformation:

The histogram is highly right-skewed because 500000 is much larger than the other income values.

After transformation:

The large value is compressed, so the distribution becomes less skewed and the values become more balanced.

The transformation does not remove the original value. Instead, it changes how the value is represented.

## Real-World Example

Consider a customer dataset containing annual income:

    20000
    25000
    30000
    35000
    40000
    1000000

The value 1000000 is much larger than the other values and can create strong right skewness.

Applying an appropriate transformation can reduce the effect of this large value.

## AI/ML Use Case

Data transformation can be useful for algorithms that are sensitive to feature distribution or feature magnitude, such as:

- Linear Regression
- Logistic Regression
- K-Nearest Neighbors
- Support Vector Machines
- Neural Networks

Tree-based algorithms such as Decision Trees and Random Forests are generally less sensitive to feature distribution.

## Important Point

Data Transformation is different from simply changing the data type.

Changing integer to float is a data type conversion.

For example:

    age = 25
    age = float(age)

Changing the mathematical representation of a feature is a data transformation.

For example:

    income = 500000
    log_income = np.log(income)

## Common Data Transformations

- Log Transformation
- Square Root Transformation
- Power Transformation
- Box-Cox Transformation
- Yeo-Johnson Transformation
- Normalization
- Standardization

## Key Takeaway

Data Transformation changes the scale, shape, or distribution of data to make it more suitable for analysis and machine learning.

The main goals are to reduce skewness, control the influence of extreme values, improve numerical stability, and make the data more suitable for certain machine learning algorithms.

```python
## Compare Before and After Transformation

import numpy as np
import matplotlib.pyplot as plt

income = [20000, 22000, 25000, 28000, 30000, 35000, 40000, 500000]

log_income = np.log(income)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.hist(income, bins=8)
plt.title("Before Transformation")
plt.xlabel("Income")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(log_income, bins=8)
plt.title("After Log Transformation")
plt.xlabel("Log Income")
plt.ylabel("Frequency")

plt.show()
```

# Log Transformation

Log Transformation is a data transformation technique that applies a logarithm to numerical values.

It is mainly used to reduce right skewness and compress very large values.

## Why Use Log Transformation?

Log Transformation is useful when:

- Data is highly right-skewed
- A feature contains very large values
- Extreme values have a strong influence
- The difference between small and large values is very high
- We want to make the distribution more balanced

## How Log Transformation Works

The basic formula is:

    y = log(x)

For example:

    log(1)     = 0
    log(10)    = 2.30
    log(100)   = 4.61
    log(1000)  = 6.91

The important idea is that the increase in large values becomes compressed.

## Simple Example

Suppose we have income data:

    income = [20000, 22000, 25000, 30000, 40000, 500000]

The value 500000 is much larger than the other values and creates right skewness.

We can apply log transformation:

    import numpy as np

    income = [20000, 22000, 25000, 30000, 40000, 500000]

    log_income = np.log(income)

    print(log_income)

## Output Explanation

The original data contains:

    20000
    22000
    25000
    30000
    40000
    500000

After log transformation, the values are approximately:

    9.90
    9.99
    10.13
    10.31
    10.60
    13.12

The value 500000 is still the largest value, but its magnitude has been compressed.



## Output Explanation

Before transformation:

The distribution is strongly right-skewed because of the very large value 500000.

After transformation:

The large value is compressed, so the distribution becomes less right-skewed and more balanced.

The histogram helps us visually compare the distribution before and after transformation.

## Log Transformation with Pandas

We can also apply log transformation directly to a Pandas column.

    import pandas as pd
    import numpy as np

    df = pd.DataFrame({
        "Income": [20000, 22000, 25000, 30000, 40000, 500000]
    })

    df["Log_Income"] = np.log(df["Income"])

    print(df)

## Important Point About Zero

The logarithm of zero is not defined.

    np.log(0)

Therefore, if a feature contains zero values, we commonly use:

    np.log1p(x)

`log1p(x)` calculates:

    log(1 + x)

For example:

    x = 0
    np.log1p(x)

Output:

    0.0

## Example with Zero Values

    import numpy as np

    values = [0, 10, 20, 100, 1000]

    transformed = np.log1p(values)

    print(transformed)

This allows us to safely transform data containing zero values.

## Important Point About Negative Values

Standard log transformation cannot be directly applied to negative values.

For example:

    np.log(-10)

produces an invalid result.

Therefore, if the feature contains negative values, another transformation such as Yeo-Johnson may be more appropriate.

## Real-World Examples

Log Transformation is commonly used for:

- Income
- House prices
- Population
- Sales
- Website visits
- Transaction amounts
- Insurance claims
- Company revenue

These variables often contain many smaller values and a few very large values.

## AI/ML Use Case

Log Transformation can be useful before models such as:

- Linear Regression
- Logistic Regression
- K-Nearest Neighbors
- Support Vector Machines
- Neural Networks

It can help reduce skewness and make relationships between variables easier for some models to learn.

## When Should We Use It?

Use Log Transformation when the feature is:

- Positively/right skewed
- Strictly positive
- Containing values that vary across a large range

Do not apply it blindly to every numerical feature.

## Key Takeaway

Log Transformation compresses large values more strongly than small values.

It is mainly used to reduce right skewness and the influence of very large values.

For positive data:

    log(x)

For data containing zero:

    log1p(x)

For data containing negative values, consider transformations such as Yeo-Johnson.

```python
## Compare Distribution Before and After Transformation

import numpy as np
import matplotlib.pyplot as plt

income = [20000, 22000, 25000, 30000, 40000, 500000]

log_income = np.log(income)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.hist(income, bins=6)
plt.title("Before Log Transformation")
plt.xlabel("Income")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(log_income, bins=6)
plt.title("After Log Transformation")
plt.xlabel("Log Income")
plt.ylabel("Frequency")

plt.show()
```

# Square Root Transformation

Square Root Transformation is a data transformation technique that applies the square root of each numerical value.

It is mainly used to reduce moderate right skewness and reduce the influence of large values.

## Why Use Square Root Transformation?

Square Root Transformation is useful when:

- Data is moderately right-skewed
- Large values have some influence on the distribution
- We want a milder transformation than log transformation
- The feature contains zero values
- The data represents counts or frequencies

## How Square Root Transformation Works

The basic formula is:

    y = √x

For example:

    √0     = 0
    √1     = 1
    √4     = 2
    √9     = 3
    √100   = 10
    √10000 = 100

The square root compresses large values while keeping zero valid.

## Simple Example

Suppose we have the number of website visits:

    visits = [10, 15, 20, 25, 30, 40, 100]

The value 100 is much larger than most other values.

We can apply square root transformation:

    import numpy as np

    visits = [10, 15, 20, 25, 30, 40, 100]

    sqrt_visits = np.sqrt(visits)

    print(sqrt_visits)

## Output Explanation

Original values:

    10
    15
    20
    25
    30
    40
    100

After square root transformation, approximately:

    3.16
    3.87
    4.47
    5.00
    5.48
    6.32
    10.00

The value 100 is still the largest value, but its magnitude has been reduced from 100 to 10.


## Output Explanation

Before transformation:

The distribution may be right-skewed because the value 100 is much larger than the other values.

After transformation:

The value 100 is compressed to 10, reducing its influence on the distribution.

The histogram allows us to visually compare the distribution before and after transformation.

## Square Root Transformation with Pandas

We can apply the transformation directly to a Pandas column.

    import pandas as pd
    import numpy as np

    df = pd.DataFrame({
        "Visits": [10, 15, 20, 25, 30, 40, 100]
    })

    df["Sqrt_Visits"] = np.sqrt(df["Visits"])

    print(df)

## Output

The DataFrame will contain both the original and transformed values:

    Visits    Sqrt_Visits
    10        3.16
    15        3.87
    20        4.47
    25        5.00
    30        5.48
    40        6.32
    100       10.00

## Important Point About Zero

Square Root Transformation can directly handle zero.

    np.sqrt(0)

Output:

    0.0

Therefore, unlike regular log transformation, square root transformation does not have a problem with zero values.

## Important Point About Negative Values

Standard square root transformation cannot be directly applied to negative values.

For example:

    np.sqrt(-10)

produces an invalid result.

Therefore, if the feature contains negative values, another transformation such as Yeo-Johnson may be more appropriate.

## Real-World Examples

Square Root Transformation can be useful for:

- Number of website visits
- Number of customer complaints
- Number of calls
- Number of transactions
- Number of accidents
- Number of defects
- Other count-based variables

## AI/ML Use Case

Square Root Transformation can be useful before machine learning models when a count-based feature is moderately right-skewed.

It can help reduce the influence of large values and make the feature distribution more balanced.

## Log vs Square Root Transformation

Log Transformation provides stronger compression of large values.

Square Root Transformation provides milder compression.

For example:

    Original value: 10000

    Square Root:
    √10000 = 100

    Log:
    log(10000) ≈ 9.21

Therefore, log transformation generally compresses large values more strongly than square root transformation.

## When Should We Use It?

Use Square Root Transformation when:

- The feature is moderately right-skewed
- The data contains zero values
- The feature represents counts or frequencies
- A milder transformation is preferred

Do not apply it blindly to every numerical feature.

## Key Takeaway

Square Root Transformation reduces the influence of large values and can help reduce moderate right skewness.

It is especially useful for count-based data and is a milder transformation compared with log transformation.

```python

## Compare Distribution Before and After Transformation

import numpy as np
import matplotlib.pyplot as plt

visits = [10, 15, 20, 25, 30, 40, 100]

sqrt_visits = np.sqrt(visits)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.hist(visits, bins=7)
plt.title("Before Square Root Transformation")
plt.xlabel("Website Visits")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(sqrt_visits, bins=7)
plt.title("After Square Root Transformation")
plt.xlabel("Square Root of Visits")
plt.ylabel("Frequency")

plt.show()
```

# Power Transformation

Power Transformation is a data transformation technique that applies a mathematical power to numerical values to make the distribution more suitable for analysis and machine learning.

It is mainly used to reduce skewness and make data more normally distributed.

## Why Use Power Transformation?

Power Transformation is useful when:

- Data is skewed
- The distribution is far from normal
- We want to reduce the effect of extreme values
- We want to stabilize variance
- We want to make the relationship between variables more suitable for modeling

## How Power Transformation Works

A power transformation applies a mathematical power to the data.

A simple example is:

    y = x^λ

Here, `λ` is called the power parameter.

Different values of `λ` produce different transformations.

For example:

    λ = 1      → No major transformation
    λ = 0.5    → Square root transformation
    λ = 2      → Square transformation
    λ = -1     → Reciprocal transformation

In practical machine learning, we often use `PowerTransformer` from Scikit-learn because it can automatically find a suitable power parameter.

## Simple Example

Suppose we have the following data:

    income = [10000, 15000, 20000, 30000, 50000, 100000, 500000]

The data is strongly right-skewed.

We can use Power Transformation to make the distribution more balanced.

## Python Example

    import pandas as pd
    from sklearn.preprocessing import PowerTransformer

    df = pd.DataFrame({
        "Income": [10000, 15000, 20000, 30000, 50000, 100000, 500000]
    })

    pt = PowerTransformer(method="yeo-johnson")

    df["Transformed_Income"] = pt.fit_transform(df[["Income"]])

    print(df)

## Output Explanation

The original `Income` column contains highly different values.

The `Transformed_Income` column contains transformed values with a more balanced distribution.

The exact transformed values depend on the power parameter selected by the transformer.

The important point is that the transformation changes the shape of the distribution rather than simply changing the data type.


## Output Explanation

Before transformation:

The income distribution is strongly right-skewed because a few values are much larger than the others.

After transformation:

The large values are compressed and the distribution becomes more balanced.

The histogram helps us compare the shape of the feature before and after transformation.

## PowerTransformer in Scikit-learn

Scikit-learn provides:

    from sklearn.preprocessing import PowerTransformer

`PowerTransformer` supports two important methods:

    PowerTransformer(method="yeo-johnson")

and

    PowerTransformer(method="box-cox")

## Yeo-Johnson Method

Yeo-Johnson can handle:

- Positive values
- Zero values
- Negative values

Example:

    values = [-10, -5, 0, 5, 10, 50, 100]

    pt = PowerTransformer(method="yeo-johnson")

    transformed = pt.fit_transform(
        pd.DataFrame(values, columns=["Value"])
    )

This makes Yeo-Johnson useful when the dataset contains zero or negative values.

## Box-Cox Method

Box-Cox is another power transformation method.

However, Box-Cox requires strictly positive values.

Example:

    values = [10, 20, 30, 50, 100, 500]

    pt = PowerTransformer(method="box-cox")

    transformed = pt.fit_transform(
        pd.DataFrame(values, columns=["Value"])
    )

Box-Cox cannot directly handle zero or negative values.

## Important Difference

Power Transformation is the broader concept.

Box-Cox and Yeo-Johnson are specific types of power transformations.

    Power Transformation
            |
            ├── Box-Cox
            |
            └── Yeo-Johnson

## Real-World Example

Suppose a company has customer spending data:

    100
    150
    200
    300
    500
    1000
    10000

Most customers spend smaller amounts, while a few customers spend very large amounts.

This creates right skewness.

A power transformation can reduce the skewness and produce a more balanced distribution.

## AI/ML Use Case

Power Transformation can be useful before algorithms such as:

- Linear Regression
- Logistic Regression
- K-Nearest Neighbors
- Support Vector Machines
- Neural Networks

It can help make the feature distribution more suitable for models that are sensitive to the distribution or scale of the input features.

## Important Point

Power Transformation does not guarantee that every dataset will become perfectly normal.

The transformation should be checked by comparing:

- Histogram
- Skewness
- Mean and median
- Q-Q plot when required

## When Should We Use It?

Use Power Transformation when:

- The numerical feature is skewed
- You want to reduce skewness
- You want a more Gaussian-like distribution
- You need a systematic transformation instead of manually choosing log or square root

## Key Takeaway

Power Transformation uses a mathematical power to change the shape of a distribution.

`PowerTransformer` in Scikit-learn can automatically estimate a suitable transformation.

The two main methods are:

- Box-Cox → requires strictly positive values
- Yeo-Johnson → can handle positive, zero, and negative values

The goal is usually to reduce skewness and make the feature distribution more suitable for statistical analysis and machine learning.

```python

## Compare Distribution Before and After Transformation

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import PowerTransformer

income = [10000, 15000, 20000, 30000, 50000, 100000, 500000]

pt = PowerTransformer(method="yeo-johnson")

transformed_income = pt.fit_transform(
    pd.DataFrame(income, columns=["Income"])
    )

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.hist(income, bins=7)
plt.title("Before Power Transformation")
plt.xlabel("Income")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(transformed_income, bins=7)
plt.title("After Power Transformation")
plt.xlabel("Transformed Income")
plt.ylabel("Frequency")

plt.show()
```

# Box-Cox Transformation

Box-Cox Transformation is a type of Power Transformation used to reduce skewness and make a numerical feature more normally distributed.

It is mainly used when the data contains only positive values.

## Why Use Box-Cox Transformation?

Box-Cox Transformation is useful when:

- Data is right-skewed
- The feature contains only positive values
- We want to reduce skewness
- We want to make the distribution more Gaussian-like
- We want to stabilize variance

## How Box-Cox Transformation Works

Box-Cox uses a parameter called lambda (`λ`) to determine the appropriate transformation.

The formula is:

    y(λ) = (x^λ - 1) / λ       when λ ≠ 0

    y(λ) = log(x)              when λ = 0

Different lambda values produce different transformations.

For example:

    λ = 1       → approximately no transformation
    λ = 0.5     → similar to square root transformation
    λ = 0       → log transformation
    λ = -1      → similar to reciprocal transformation

In practice, we usually allow the algorithm to find a suitable lambda value automatically.

## Important Requirement

Box-Cox requires strictly positive values.

For example:

    [10, 20, 30, 50, 100]

is valid.

But:

    [0, 10, 20, 30]

is not valid.

And:

    [-10, 10, 20, 30]

is also not valid.

If the data contains zero or negative values, Yeo-Johnson Transformation is a better choice.

## Simple Example

Suppose we have house prices:

    prices = [100, 120, 150, 180, 250, 400, 1000]

The value `1000` is much larger than most of the other values, creating right skewness.

We can apply Box-Cox Transformation.



## Output Explanation

The `boxcox()` function returns two things:

    transformed_prices

This contains the transformed values.

    lambda_value

This contains the lambda value selected by the algorithm.

The algorithm chooses a lambda that attempts to make the distribution more suitable for normality.

The exact values may vary depending on the input data.



## Output Explanation

Before transformation:

The distribution is right-skewed because the value `1000` is much larger than the other values.

After transformation:

The large values are compressed and the distribution generally becomes less skewed.

The histogram allows us to visually compare the original and transformed distributions.

## Checking Skewness

We can calculate skewness before and after transformation.

    from scipy.stats import skew

    before = skew(prices)
    after = skew(transformed_prices)

    print("Skewness Before:", before)
    print("Skewness After:", after)

If the transformation is useful, the absolute skewness will generally become smaller.

For example:

    Skewness Before: 2.5
    Skewness After: 0.3

This indicates that the distribution became considerably less skewed.

The exact values depend on the dataset.

## Box-Cox Using Scikit-learn

We can also use Scikit-learn's `PowerTransformer`.

    import pandas as pd
    from sklearn.preprocessing import PowerTransformer

    df = pd.DataFrame({
        "Price": [100, 120, 150, 180, 250, 400, 1000]
    })

    pt = PowerTransformer(method="box-cox")

    df["BoxCox_Price"] = pt.fit_transform(df[["Price"]])

    print(df)

This automatically estimates a suitable lambda value.

## Real-World Examples

Box-Cox Transformation can be useful for:

- House prices
- Income
- Sales amount
- Revenue
- Insurance claims
- Transaction amounts
- Product prices
- Waiting times

These features can contain positive values with strong right skewness.

## AI/ML Use Case

Box-Cox Transformation can be useful before models such as:

- Linear Regression
- Logistic Regression
- K-Nearest Neighbors
- Support Vector Machines
- Neural Networks

It can help make the distribution more balanced and can improve the suitability of the feature for algorithms that are sensitive to feature distribution.

## Box-Cox vs Log Transformation

Log Transformation always uses a fixed logarithmic transformation.

Box-Cox is more flexible because it finds a suitable lambda value.

For example:

    Log Transformation:
    y = log(x)

    Box-Cox:
    y = transformation(x, λ)

If the best lambda is close to `0`, the Box-Cox transformation behaves similarly to a log transformation.

## Box-Cox vs Yeo-Johnson

Box-Cox:

- Requires strictly positive values
- Cannot directly handle zero
- Cannot directly handle negative values

Yeo-Johnson:

- Can handle positive values
- Can handle zero
- Can handle negative values

Therefore, if the feature contains zero or negative values, Yeo-Johnson is usually more appropriate.

## When Should We Use Box-Cox?

Use Box-Cox when:

- The feature is numerical
- The feature contains only positive values
- The distribution is skewed
- You want to reduce skewness
- You want a more Gaussian-like distribution

## Key Takeaway

Box-Cox Transformation is a flexible power transformation for positive numerical data.

It automatically finds a suitable lambda value to transform the data and reduce skewness.

Remember:

    Box-Cox → Positive values only

    Yeo-Johnson → Positive, zero, and negative values

```python
## Python Example

import numpy as np
from scipy.stats import boxcox

prices = np.array([100, 120, 150, 180, 250, 400, 1000])
transformed_prices, lambda_value = boxcox(prices)

print("Lambda:", lambda_value)
print("Transformed Data:", transformed_prices)
```

```python
## Compare Distribution Before and After Transformation

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import boxcox

prices = np.array([100, 120, 150, 180, 250, 400, 1000])

transformed_prices, lambda_value = boxcox(prices)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.hist(prices, bins=7)
plt.title("Before Box-Cox Transformation")
plt.xlabel("House Price")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(transformed_prices, bins=7)
plt.title("After Box-Cox Transformation")
plt.xlabel("Transformed Price")
plt.ylabel("Frequency")

plt.show()
```

# Yeo-Johnson Transformation

Yeo-Johnson Transformation is a type of Power Transformation used to reduce skewness and make numerical data more suitable for statistical analysis and machine learning.

The main advantage of Yeo-Johnson is that it can handle positive values, zero values, and negative values.

## Why Use Yeo-Johnson Transformation?

Yeo-Johnson Transformation is useful when:

- Data is skewed
- The feature contains zero values
- The feature contains negative values
- We want to reduce skewness
- We want to make the distribution more Gaussian-like
- We want to stabilize variance

## Important Advantage

Unlike Box-Cox Transformation, Yeo-Johnson does not require the data to be strictly positive.

For example, this data can be transformed:

    values = [-100, -50, -10, 0, 10, 50, 100]

This is one of the main reasons Yeo-Johnson is useful in real-world datasets.

## How Yeo-Johnson Works

Yeo-Johnson uses a parameter called lambda (`λ`) to determine the appropriate transformation.

The algorithm searches for a suitable lambda value that helps make the distribution more balanced.

You normally do not need to manually calculate the formula because Scikit-learn can estimate the appropriate lambda automatically.

## Simple Example

Suppose we have profit data:

    profit = [-500, -200, -50, 0, 100, 300, 1000]

The dataset contains:

- Negative values
- Zero
- Positive values

A normal Box-Cox transformation cannot directly handle this data.

Yeo-Johnson can handle it.



## Output Explanation

The original `Profit` column contains negative, zero, and positive values.

The `Transformed_Profit` column contains transformed values.

The transformation changes the distribution of the feature while preserving the order of the observations.

The exact transformed values depend on the data and the lambda value estimated by the transformer.


## Output Explanation

Before transformation:

The distribution may be skewed because some values are much larger or smaller than the others.

After transformation:

The values are transformed to produce a more balanced distribution.

The histogram allows us to visually compare the shape of the data before and after transformation.

## Checking Skewness

We can calculate skewness before and after transformation.

    from scipy.stats import skew

    before = skew(profit)
    after = skew(transformed_profit.ravel())

    print("Skewness Before:", before)
    print("Skewness After:", after)

If the transformation works well, the absolute skewness will generally become smaller.

For example:

    Skewness Before: 1.8
    Skewness After: 0.2

The exact values depend on the dataset.

## Using PowerTransformer

Yeo-Johnson is available through Scikit-learn:

    from sklearn.preprocessing import PowerTransformer

    pt = PowerTransformer(method="yeo-johnson")

    transformed = pt.fit_transform(df[["Profit"]])

`fit_transform()` performs two operations:

    fit → learns the appropriate transformation parameters from the data

    transform → applies the learned transformation to the data

## Transforming Training and Testing Data

In machine learning, we should fit the transformer only on the training data.

    pt = PowerTransformer(method="yeo-johnson")

    X_train_transformed = pt.fit_transform(X_train)

    X_test_transformed = pt.transform(X_test)

We use `fit_transform()` on training data because the transformer needs to learn the transformation parameters.

We use only `transform()` on test data so that information from the test set does not influence the transformation.

## Real-World Examples

Yeo-Johnson Transformation can be useful for:

- Profit and loss
- Temperature differences
- Financial changes
- Customer balances
- Measurement data
- Sensor readings
- Business metrics
- Features containing positive and negative values

## AI/ML Use Case

Yeo-Johnson can be useful before:

- Linear Regression
- Logistic Regression
- K-Nearest Neighbors
- Support Vector Machines
- Neural Networks

It can help reduce skewness and make numerical features more suitable for models that are sensitive to feature distribution.

## Yeo-Johnson vs Box-Cox

Box-Cox:

    Positive values only

Yeo-Johnson:

    Positive values
    Zero values
    Negative values

Example:

    Data = [10, 20, 30, 40]

Both Box-Cox and Yeo-Johnson can be used.

    Data = [-10, 0, 10, 20]

Yeo-Johnson can be used, but Box-Cox cannot be directly applied.

## Important Point

Yeo-Johnson does not guarantee a perfectly normal distribution.

After transformation, always check the result using:

- Histogram
- Skewness
- Mean and median
- Q-Q plot when required

## When Should We Use It?

Use Yeo-Johnson when:

- The numerical feature is skewed
- The feature contains zero values
- The feature contains negative values
- You want to reduce skewness
- You want a more Gaussian-like distribution

## Key Takeaway

Yeo-Johnson Transformation is a flexible Power Transformation that can handle positive, zero, and negative values.

It automatically learns a suitable transformation parameter and is especially useful when Box-Cox cannot be applied.

    Box-Cox → Positive values only

    Yeo-Johnson → Positive + Zero + Negative values

```python
## Python Example

import pandas as pd
from sklearn.preprocessing import PowerTransformer

df = pd.DataFrame({
"Profit": [-500, -200, -50, 0, 100, 300, 1000]
    })

pt = PowerTransformer(method="yeo-johnson")

df["Transformed_Profit"] = pt.fit_transform(df[["Profit"]])

print(df)
```

# Handling Skewed Data

Skewed Data is numerical data where the values are not distributed symmetrically around the center.

In a perfectly symmetric distribution, the left and right sides have a similar shape.

In skewed data, one side has a longer or heavier tail.

## Types of Skewness

There are two main types of skewness:

1. Right Skewness
2. Left Skewness

## Right Skewness

Right-skewed data has a long tail toward larger values.

It is also called positive skewness.

Example:

    [10, 12, 13, 15, 18, 20, 100]

Most values are small, but a few values are very large.

Common examples include:

- Income
- House prices
- Sales
- Revenue
- Website visits

## Left Skewness

Left-skewed data has a long tail toward smaller values.

It is also called negative skewness.

Example:

    [1, 80, 85, 88, 90, 92, 95]

Most values are high, but a few values are very low.

## Why Is Skewness Important?

Highly skewed data can affect some statistical methods and machine learning algorithms.

It can:

- Affect the mean
- Make the distribution difficult to model
- Influence statistical relationships
- Reduce the effectiveness of some algorithms
- Make some assumptions of statistical models less suitable

Not every ML algorithm requires normally distributed features, so skewness should not be removed blindly.

## Detecting Skewness

We can use Pandas to calculate skewness.

    import pandas as pd

    df = pd.DataFrame({
        "Income": [20000, 22000, 25000, 30000, 35000, 50000, 500000]
    })

    print(df["Income"].skew())

## Output Explanation

If the skewness value is:

    0

the distribution is approximately symmetric.

If the value is:

    Positive

the distribution is right-skewed.

If the value is:

    Negative

the distribution is left-skewed.

A larger absolute skewness value generally indicates stronger skewness.

There is no single universal cutoff that perfectly determines when data is too skewed, so the distribution should also be inspected visually and in context.

## Visualizing Skewed Data

    import matplotlib.pyplot as plt

    plt.hist(df["Income"], bins=7)
    plt.title("Skewed Income Distribution")
    plt.xlabel("Income")
    plt.ylabel("Frequency")
    plt.show()

The histogram helps us see whether the data has a long tail on the left or right side.

## Handling Right-Skewed Data

Right-skewed data can often be handled using transformations such as:

- Log Transformation
- Square Root Transformation
- Box-Cox Transformation
- Yeo-Johnson Transformation
- Power Transformation

For example:

    import numpy as np

    df["Log_Income"] = np.log(df["Income"])

This compresses large values and can reduce right skewness.

## Handling Zero Values

If the feature contains zero values, use:

    np.log1p(x)

instead of:

    np.log(x)

Example:

    values = [0, 10, 20, 50, 100]

    transformed = np.log1p(values)

`log1p(x)` calculates:

    log(1 + x)

Therefore, zero can safely be transformed.

## Handling Negative Values

Standard log and Box-Cox transformations cannot directly handle negative values.

For example:

    values = [-100, -50, 0, 50, 100]

In this situation, Yeo-Johnson can be used.

    from sklearn.preprocessing import PowerTransformer

    pt = PowerTransformer(method="yeo-johnson")

    transformed = pt.fit_transform(
        pd.DataFrame(values, columns=["Value"])
    )

## Output Explanation

Before transformation:

The distribution has a long right tail because of the large income value.

After transformation:

The large values are compressed, making the distribution less right-skewed.

The histogram allows us to visually compare the shape of the distribution before and after transformation.

## Checking Skewness Before and After

    from scipy.stats import skew

    before = skew(income)
    after = skew(log_income)

    print("Skewness Before:", before)
    print("Skewness After:", after)

If the absolute skewness becomes smaller after transformation, the distribution has generally become more symmetric.

For example:

    Skewness Before: 2.5
    Skewness After: 0.4

The exact values depend on the dataset.

## Important Point About Outliers

Skewness and outliers are related but they are not the same thing.

An outlier is an unusually extreme observation.

Skewness describes the overall asymmetry of the distribution.

Therefore, do not automatically remove extreme values just because the data is skewed.

Sometimes the extreme value is a valid observation.

For example:

    Employee salaries:
    25000
    30000
    35000
    40000
    500000

The ₹5,00,000 salary may be a valid salary rather than an error.

In such cases, transformation may be more appropriate than deleting the observation.

## Real-World Example

Consider an e-commerce dataset containing customer spending:

    500
    700
    900
    1200
    1500
    2000
    50000

Most customers spend relatively small amounts, while one customer spends ₹50,000.

The feature is strongly right-skewed.

A log or power transformation can reduce the effect of the large value while keeping the original observation.

## AI/ML Use Case

Handling skewness can be useful for:

- Linear Regression
- Logistic Regression
- K-Nearest Neighbors
- Support Vector Machines
- Neural Networks

Tree-based algorithms such as Decision Trees and Random Forests are generally much less sensitive to skewed feature distributions.

## General Process for Handling Skewed Data

    1. Identify the numerical feature
    2. Calculate skewness
    3. Visualize the distribution
    4. Identify whether the data is left- or right-skewed
    5. Check whether extreme values are valid
    6. Select an appropriate transformation
    7. Apply the transformation
    8. Compare the distribution before and after
    9. Recalculate skewness
    10. Use the transformed feature when appropriate

## Key Takeaway

Skewed data is data that is not symmetrically distributed.

Right-skewed data has a long tail toward larger values.

Left-skewed data has a long tail toward smaller values.

Common methods for handling skewness include Log, Square Root, Power, Box-Cox, and Yeo-Johnson transformations.

Always compare the distribution before and after transformation and check whether the transformation actually improves the data for the intended analysis or ML model.

# Normalization

Normalization is a data scaling technique that transforms numerical values to a specific range, usually between 0 and 1.

It is commonly used when features have very different ranges and the machine learning algorithm is sensitive to feature magnitude.

## Why Use Normalization?

Suppose a dataset contains:

    Age       → 20 to 60
    Salary    → 20,000 to 1,00,000
    Experience → 1 to 20

Salary has much larger numerical values than Age and Experience.

Some algorithms may give more importance to Salary simply because its numerical scale is larger.

Normalization puts the values into a common range.

## Min-Max Normalization

The most common normalization technique is Min-Max Scaling.

The formula is:

    x_scaled = (x - min(x)) / (max(x) - min(x))

This converts the values into a range between 0 and 1.

## Simple Example

Suppose we have:

    Age = [20, 30, 40, 50, 60]

Minimum value:

    20

Maximum value:

    60

For Age = 30:

    x_scaled = (30 - 20) / (60 - 20)
             = 10 / 40
             = 0.25

Therefore:

    30 → 0.25

## Python Example

    import numpy as np

    age = np.array([20, 30, 40, 50, 60])

    normalized_age = (age - age.min()) / (age.max() - age.min())

    print(normalized_age)

## Output

    [0.   0.25 0.5  0.75 1.  ]

The smallest value becomes `0`.

The largest value becomes `1`.

The values between them are scaled proportionally.


## Output Explanation

Both features are converted to values between 0 and 1.

The original ranges were:

    Age       → 20 to 60
    Salary    → 20000 to 100000

After normalization:

    Age       → 0 to 1
    Salary    → 0 to 1

The original scale is changed, but the relative ordering of the values is preserved.

## Important Point About Distribution

Normalization changes the scale of the data but does not normally change the overall shape of the distribution.

For example, if the original feature is right-skewed, Min-Max normalization will generally keep it right-skewed.

It changes:

    20000 → 0
    40000 → 0.25
    60000 → 0.50
    80000 → 0.75
    100000 → 1

It does not automatically remove skewness.

Therefore:

    Transformation → changes distribution shape

    Normalization → changes scale/range

## Real-World Example

Consider an employee dataset:

    Age          → 20 to 60
    Salary       → 20000 to 200000
    Experience   → 1 to 30

Without scaling, Salary has much larger numerical values.

Normalization converts all features into a common range:

    Age          → 0 to 1
    Salary       → 0 to 1
    Experience   → 0 to 1

This prevents the larger numerical scale of Salary from dominating distance-based calculations.

## AI/ML Use Case

Normalization is commonly useful for:

- K-Nearest Neighbors
- K-Means Clustering
- Support Vector Machines
- Neural Networks
- Gradient-based algorithms

For example, KNN calculates distances between observations.

If Salary ranges from 20,000 to 2,00,000 and Age ranges from 20 to 60, Salary can dominate the distance calculation.

Normalization puts them on a comparable scale.

## Important Point About Training and Testing Data

Never fit the scaler separately on training and testing data.

Correct approach:

    scaler = MinMaxScaler()

    X_train_scaled = scaler.fit_transform(X_train)

    X_test_scaled = scaler.transform(X_test)

The scaler learns the minimum and maximum values from the training data.

Then the same scaling parameters are applied to the test data.

## Normalization vs Transformation

Normalization changes the range of values.

Transformation changes the mathematical representation or shape of the distribution.

Example:

    Original:
    [10, 20, 30, 100]

    Log Transformation:
    [2.30, 3.00, 3.40, 4.61]

    Min-Max Normalization:
    [0.00, 0.11, 0.22, 1.00]

The log transformation can reduce skewness.

Min-Max normalization only scales the values to a specified range.

## When Should We Use Normalization?

Use normalization when:

- Features have different numerical ranges
- The algorithm is sensitive to feature scale
- You want values within a fixed range
- Distance calculations are involved
- Neural networks or gradient-based algorithms are being used

## When Should We Be Careful?

Min-Max normalization is sensitive to outliers.

For example:

    [10, 20, 30, 40, 10000]

The value `10000` becomes the maximum.

Most other values will then be compressed very close to zero.

In such situations, RobustScaler or another appropriate scaling approach may be better.

## Key Takeaway

Normalization scales numerical values to a fixed range, commonly 0 to 1.

The most common method is Min-Max Scaling.

Normalization changes the scale of the feature but does not normally remove skewness.

Remember:

    Transformation → changes distribution

    Normalization → changes range

    Standardization → changes values based on mean and standard deviation

```python
## Using MinMaxScaler

#Scikit-learn provides `MinMaxScaler` for normalization.

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.DataFrame({
    "Age": [20, 30, 40, 50, 60],
        "Salary": [20000, 40000, 60000, 80000, 100000]
    })

scaler = MinMaxScaler()

df_normalized = scaler.fit_transform(df)

print(df_normalized)
```

```python

## Compare Before and After Scaling

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

df = pd.DataFrame({
    "Age": [20, 30, 40, 50, 60],
    "Salary": [20000, 40000, 60000, 80000, 100000]
    })

scaler = MinMaxScaler()

df_normalized = pd.DataFrame(
    scaler.fit_transform(df),
    columns=df.columns
    )

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.hist(df["Salary"], bins=5)
plt.title("Before Normalization")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(df_normalized["Salary"], bins=5)
plt.title("After Normalization")
plt.xlabel("Normalized Salary")
plt.ylabel("Frequency")

plt.show()
```

# Standardization

Standardization is a feature scaling technique that transforms numerical data so that the feature has a mean of 0 and a standard deviation of 1.

It is also called Z-score Standardization.

## Why Use Standardization?

Standardization is useful when numerical features have different scales.

For example:

    Age         → 20 to 60
    Salary      → 20000 to 200000
    Experience  → 1 to 30

Salary has much larger numerical values than Age and Experience.

Standardization puts the features onto a comparable scale.

## How Standardization Works

The formula is:

    z = (x - μ) / σ

Where:

    x  = original value
    μ  = mean of the feature
    σ  = standard deviation of the feature
    z  = standardized value

After standardization:

    Mean ≈ 0
    Standard Deviation ≈ 1

## Simple Example

Suppose we have:

    Age = [20, 30, 40, 50, 60]

The mean is:

    Mean = 40

For Age = 30:

    z = (30 - 40) / σ

The value becomes negative because 30 is below the mean.

For Age = 50:

    z = (50 - 40) / σ

The value becomes positive because 50 is above the mean.

The exact value depends on the standard deviation.

## Python Example

    import numpy as np

    age = np.array([20, 30, 40, 50, 60])

    mean = age.mean()
    std = age.std()

    standardized_age = (age - mean) / std

    print(standardized_age)

## Output

The result will be approximately:

    [-1.41, -0.71, 0.00, 0.71, 1.41]

## Output Explanation

The standardized values tell us how far each value is from the mean in terms of standard deviations.

For example:

    -1.41 → 1.41 standard deviations below the mean
    -0.71 → 0.71 standard deviations below the mean
     0.00 → exactly at the mean
     0.71 → 0.71 standard deviations above the mean
     1.41 → 1.41 standard deviations above the mean



## Output Explanation

Both Age and Salary are transformed into standardized values.

The original ranges were very different:

    Age       → 20 to 60
    Salary    → 20000 to 100000

After standardization, both features are centered around zero and have a standard deviation close to one.

## Important Point About Distribution

Standardization changes the scale and location of the data.

It does not normally remove skewness.

For example, if the original data is right-skewed, standardization will generally keep the same right-skewed shape.

It changes the values so that:

    Mean ≈ 0
    Standard Deviation ≈ 1

Therefore:

    Transformation → can change distribution shape

    Standardization → changes scale and centers data around zero

## Real-World Example

Consider an employee dataset:

    Age          → 20 to 60
    Salary       → 20000 to 200000
    Experience   → 1 to 30

These features have very different scales.

Standardization converts them into comparable standardized values.

This is useful when an ML algorithm calculates distances, similarities, or uses optimization based on feature values.

## AI/ML Use Case

Standardization is commonly useful for:

- Linear Regression
- Logistic Regression
- K-Nearest Neighbors
- K-Means Clustering
- Support Vector Machines
- Principal Component Analysis
- Neural Networks
- Gradient-based algorithms

For example, KNN calculates distances between observations.

Without scaling, Salary could dominate the distance because its numerical values are much larger.

Standardization reduces this scale difference.

## Important Point About Training and Testing Data

The scaler should be fitted only on the training data.

Correct approach:

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(X_train)

    X_test_scaled = scaler.transform(X_test)

`fit_transform()` learns the mean and standard deviation from the training data and transforms it.

`transform()` uses the already learned values to transform the test data.

This prevents information from the test set from influencing the scaling process.

## Standardization vs Normalization

Normalization usually scales values to a fixed range such as 0 to 1.

Standardization centers values around 0 and scales them based on standard deviation.

Example:

    Normalization:
    10 → 0.00
    20 → 0.25
    30 → 0.50
    40 → 0.75
    50 → 1.00

    Standardization:
    10 → -1.41
    20 → -0.71
    30 →  0.00
    40 →  0.71
    50 →  1.41

Normalization has a fixed range.

Standardization does not have a fixed minimum and maximum range.

## Standardization vs Transformation

Transformation changes the mathematical representation and can change the shape of a distribution.

Standardization changes the location and scale.

For example:

    Original:
    [10, 20, 30, 100]

    Log Transformation:
    [2.30, 3.00, 3.40, 4.61]

    Standardization:
    Values are converted into z-scores based on their mean and standard deviation.

A log transformation may reduce skewness.

Standardization does not normally remove skewness.

## Important Point About Outliers

Standardization is sensitive to outliers because the mean and standard deviation are affected by extreme values.

For example:

    [10, 20, 30, 40, 10000]

The value `10000` can greatly affect the mean and standard deviation.

In such situations, RobustScaler may be more appropriate.

## When Should We Use Standardization?

Use standardization when:

- Features have different scales
- The ML algorithm is sensitive to feature magnitude
- The algorithm uses distances
- The algorithm uses gradient-based optimization
- PCA is being performed
- You want features centered around zero

## Key Takeaway

Standardization converts numerical features so that they have approximately:

    Mean = 0
    Standard Deviation = 1

The formula is:

    z = (x - μ) / σ

Standardization changes the scale and centers the data but does not normally remove skewness.

Remember:

    Transformation → changes distribution shape

    Normalization → scales values to a fixed range

    Standardization → centers values around 0 and scales by standard deviation

```python
## Using StandardScaler

#Scikit-learn provides `StandardScaler`.

import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "Age": [20, 30, 40, 50, 60],
        "Salary": [20000, 40000, 60000, 80000, 100000]
    })

scaler = StandardScaler()

df_standardized = scaler.fit_transform(df)

print(df_standardized)
```

```python
## Python Example

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew

income = np.array([
        20000, 22000, 25000, 30000,
        35000, 50000, 500000
    ])


log_income = np.log(income)

print("Skewness Before:", skew(income))
print("Skewness After:", skew(log_income))
```
