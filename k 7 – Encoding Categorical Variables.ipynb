# What is Categorical Encoding?

Categorical Encoding is the process of converting categorical values into numerical values so that Machine Learning algorithms can understand and process them.

Machine Learning models generally work with numbers, but real-world datasets often contain categorical values such as `"Male"`, `"Female"`, `"Chennai"`, `"Bangalore"`, `"Low"`, `"Medium"`, and `"High"`.


These values are categorical, so they may need to be converted into numerical representations before being given to a Machine Learning model.

For example:

```text
Chennai   → 0
Bangalore → 1
Mumbai    → 2
```

However, we should not blindly assign numbers to categories because the numbers may create a false relationship between categories.

For example, if we encode:

```text
Chennai   → 0
Bangalore → 1
Mumbai    → 2
```

the model might interpret `Mumbai (2)` as greater than `Bangalore (1)` and `Bangalore (1)` as greater than `Chennai (0)`. But cities do not have this type of natural order.

Therefore, the encoding technique should be selected based on the type of categorical variable.

Common approaches include:

* Nominal variables → One-Hot Encoding
* Ordinal variables → Ordinal Encoding
* Binary variables → Binary or Label Encoding
* High-cardinality variables → Frequency Encoding or Target Encoding

Categorical Encoding is important because it:

* Converts categorical data into a machine-readable format.
* Allows categorical features to be used by Machine Learning algorithms.
* Makes categorical data suitable for mathematical operations.
* Helps prepare real-world data for model training.

Example:

```text
Before Encoding:

Gender
------
Male
Female
Female
Male

After Encoding:

Gender
------
0
1
1
0
```

The main idea is:

```text
Categorical Data
       ↓
Choose suitable encoding technique
       ↓
Numerical Representation
       ↓
Machine Learning Model
```

The most important rule is: **Do not choose an encoding technique blindly. The encoding method should match the meaning and structure of the categorical variable.**


````python

import pandas as pd

df = pd.DataFrame({
    "City": ["Chennai", "Bangalore", "Chennai", "Mumbai"],
    "Experience_Level": ["Junior", "Senior", "Mid", "Senior"]
})

print(df)
````

# Nominal Variables

Nominal variables are categorical variables whose categories **do not have any natural order or ranking**.

Each category simply represents a different group or name. One category is not greater or smaller than another.

Examples:

```text
City:
Chennai
Bangalore
Mumbai
Delhi
```

There is no meaningful order such as:

```text
Chennai < Bangalore < Mumbai
```

Other examples of nominal variables are:

```text
Gender:
Male
Female

Color:
Red
Blue
Green

Department:
HR
IT
Finance
Sales

Payment_Method:
Cash
Card
UPI
```

For example:

Since these categories have **no natural ranking**, assigning numbers directly can create a false relationship.

For example:

```text
Chennai   → 0
Bangalore → 1
Mumbai    → 2
```

A Machine Learning model could incorrectly interpret `Mumbai (2)` as being greater than `Bangalore (1)`.

For nominal variables, **One-Hot Encoding is usually a suitable choice** because it represents each category separately without creating an artificial order.

Example:

```text
City_Chennai  City_Bangalore  City_Mumbai
1             0               0
0             1               0
0             0               1
1             0               0
```

When to use nominal variables:

* City names
* Product categories
* Departments
* Colors
* Countries
* Payment methods
* Browser types

The key point is: **Nominal = categories with names but no meaningful order.**


````python

import pandas as pd

df = pd.DataFrame({
    "City": ["Chennai", "Bangalore", "Mumbai", "Chennai"],
    "Department": ["IT", "HR", "Finance", "IT"]
})

print(df)
````

# Ordinal Variables

Ordinal variables are categorical variables whose categories have a **meaningful order or ranking**.

Unlike nominal variables, ordinal categories can be arranged from lower to higher, worse to better, or smaller to larger.

Examples:

```text
Experience Level:
Junior
Mid
Senior
```

Here, there is a natural order:

```text
Junior < Mid < Senior
```

Other examples:

```text
Satisfaction:
Poor
Average
Good
Excellent

Education:
School
Bachelor's
Master's
PhD

Size:
Small
Medium
Large

Risk:
Low
Medium
High
```

For example:

Since `Junior`, `Mid`, and `Senior` have a meaningful order, we can represent them numerically:

```text
Junior → 0
Mid    → 1
Senior → 2
```

This is different from nominal data because the numerical values represent the **ranking** of the categories.

For ordinal variables, **Ordinal Encoding** is commonly used.

Example:

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(categories=[["Junior", "Mid", "Senior"]])

df["Experience_Encoded"] = encoder.fit_transform(
    df[["Experience_Level"]]
)

print(df)
```

Output:

```text
  Employee Experience_Level  Experience_Encoded
0        A              Junior                 0.0
1        B              Senior                 2.0
2        C                 Mid                 1.0
3        D              Senior                 2.0
```

When to use ordinal variables:

* Education level
* Customer satisfaction
* Product size
* Experience level
* Risk level
* Performance rating
* Service quality

The key point is: **Ordinal = categories with a meaningful order or ranking.**


````python

import pandas as pd

df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D"],
    "Experience_Level": ["Junior", "Senior", "Mid", "Senior"]
})

print(df)

````

# Binary Variables

Binary variables are categorical variables that have only **two possible categories or values**.

Examples:

```text id="k3c8yq"
Gender:
Male
Female

Yes/No:
Yes
No

Account Status:
Active
Inactive

Loan Approval:
Approved
Rejected

Subscription:
Subscribed
Not Subscribed
```

For example:

Since there are only two categories, they can usually be represented using two numerical values:

```text id="8xqj4v"
Yes → 1
No  → 0
```

We can use `map()` to perform this encoding:

```python id="0m6wqj"
df["Subscribed_Encoded"] = df["Subscribed"].map({
    "No": 0,
    "Yes": 1
})

print(df)
```

Output:

```text id="8q4n8m"
  Customer Subscribed  Subscribed_Encoded
0        A        Yes                   1
1        B         No                   0
2        C        Yes                   1
3        D         No                    0
```

For binary variables, **binary encoding or simple 0/1 mapping is usually appropriate** because there are only two categories.

The important thing is to decide which category should represent `0` and which should represent `1`. For example:

```text id="2qgrk0"
Approved → 1
Rejected → 0
```

This makes sense when `1` represents the positive or active condition.

However, the choice depends on the business meaning. For example:

```text id="h7v4os"
Active   → 1
Inactive → 0
```

When to use binary variables:

* Yes/No
* True/False
* Active/Inactive
* Approved/Rejected
* Present/Absent
* Success/Failure

The key point is: **Binary = exactly two categories, and they can usually be represented as 0 and 1.**


````python
import pandas as pd

df = pd.DataFrame({
    "Customer": ["A", "B", "C", "D"],
    "Subscribed": ["Yes", "No", "Yes", "No"]
})

print(df)
````

# Label Encoding

Label Encoding is a technique that converts each categorical value into a **unique integer value**.

For example:

```text
Male   → 0
Female → 1
```

or:

```text
Chennai   → 0
Bangalore → 1
Mumbai    → 2
```

We can use `LabelEncoder` from Scikit-learn.


The exact numbers are assigned based on the encoder's ordering of the categories. The important point is that each category receives a unique number.

We can also check which category corresponds to each encoded value:

```python
print(encoder.classes_)
```

Output:

```text
['Bangalore' 'Chennai' 'Mumbai']
```

So:

```text
Bangalore → 0
Chennai   → 1
Mumbai    → 2
```

Label Encoding is most appropriate when the categorical variable is **ordinal**, where the numerical order represents a real ranking.

Example:

```text
Junior → 0
Mid    → 1
Senior → 2
```

It can also be useful for a **binary variable**:

```text
No  → 0
Yes → 1
```

However, Label Encoding is generally **not appropriate for nominal variables with multiple categories** such as City, Color, or Department.

For example:

```text
Chennai   → 0
Bangalore → 1
Mumbai    → 2
```

The model may interpret `Mumbai > Bangalore > Chennai`, even though cities have no natural order.

When to use Label Encoding:

* Ordinal categorical variables
* Binary categorical variables
* Some tree-based Machine Learning models where integer category representations are handled appropriately

When not to use it:

* Nominal variables with multiple categories when the model could interpret the numbers as ordered values

The key point is: **Label Encoding assigns an integer label to each category, but those numbers can create an artificial order if the categories are not naturally ordered.**


````python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    "City": ["Chennai", "Bangalore", "Mumbai", "Chennai"]
})

encoder = LabelEncoder()

df["City_Encoded"] = encoder.fit_transform(df["City"])

print(df)
````

# Ordinal Encoding

Ordinal Encoding is a technique used to convert **ordinal categorical variables** into numerical values while preserving their natural order.

Ordinal variables have categories with a meaningful ranking.

For example:

```text
Experience Level:
Junior
Mid
Senior
```

The natural order is:

```text
Junior < Mid < Senior
```

We can represent this order numerically:

```text
Junior → 0
Mid    → 1
Senior → 2
```



Here, we explicitly provide the category order:

```python
categories=[["Junior", "Mid", "Senior"]]
```

This is important because the encoder needs to know the **correct business order**.

Another example:

```text
Customer Satisfaction:

Poor      → 0
Average   → 1
Good      → 2
Excellent → 3
```

The numbers represent the ranking, so the model can understand that `Excellent` is higher than `Good`, and `Good` is higher than `Average`.

When to use Ordinal Encoding:

* Education level
* Satisfaction level
* Risk level
* Experience level
* Performance rating
* Product size
* Service quality

Do not use Ordinal Encoding for nominal variables such as:

```text
City:
Chennai
Bangalore
Mumbai
```

There is no natural order between these cities, so assigning:

```text
Chennai   → 0
Bangalore → 1
Mumbai    → 2
```

would create an artificial ranking.

The key point is: **Ordinal Encoding should be used when categories have a meaningful and known order.**


````python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    "Experience_Level": ["Junior", "Senior", "Mid", "Senior"]
})

encoder = OrdinalEncoder(
    categories=[["Junior", "Mid", "Senior"]]
)

df["Experience_Encoded"] = encoder.fit_transform(
    df[["Experience_Level"]]
)

print(df)
````

# One-Hot Encoding

One-Hot Encoding is a technique used to convert categorical values into **separate binary columns** containing `0` and `1`.

It is commonly used for **nominal variables**, where categories do not have any natural order.

For example:

```text id="z2sm5p"
City:
Chennai
Bangalore
Mumbai
```

There is no natural order between these cities, so assigning:

```text id="b5u6p8"
Chennai   → 0
Bangalore → 1
Mumbai    → 2
```

could create an incorrect relationship between the categories.

Instead, One-Hot Encoding creates one column for each category:

```text id="o6j0b8"
City        Chennai  Bangalore  Mumbai
Chennai        1         0         0
Bangalore      0         1         0
Mumbai         0         0         1
Chennai        1         0         0
```

Each row has `1` for the category it belongs to and `0` for the other categories.

Example using Pandas:

We can also get `0` and `1` instead of `True` and `False`:

One-Hot Encoding can also be performed using Scikit-learn:

```python id="3j9x0n"
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False)

result = encoder.fit_transform(df[["City"]])

print(result)
```

One-Hot Encoding is useful for:

* City
* Department
* Color
* Country
* Product category
* Payment method
* Browser type

However, if a categorical feature contains thousands of unique categories, One-Hot Encoding can create thousands of columns. This is called the **high-cardinality problem** and can increase memory usage, computation time, and model complexity.

For binary variables, One-Hot Encoding can also create two columns, but usually a single `0/1` column is more efficient.

The key point is: **One-Hot Encoding represents each category as a separate binary feature without creating an artificial numerical order.**


````python

import pandas as pd

df = pd.DataFrame({
    "City": ["Chennai", "Bangalore", "Mumbai", "Chennai"]
})

encoded_df = pd.get_dummies(df, columns=["City"])

print(encoded_df)
````

# Dummy Variables

Dummy Variables are binary variables created from categorical variables to represent whether a particular category is present or not.

They are closely related to **One-Hot Encoding**.

For example, suppose we have:

```text
City:
Chennai
Bangalore
Mumbai
```

One-Hot Encoding creates:

```text
City        Chennai  Bangalore  Mumbai
Chennai        1         0         0
Bangalore      0         1         0
Mumbai         0         0         1
```

With Dummy Variables, we can remove one category and use it as a **reference category**:

```text
City        Bangalore  Mumbai
Chennai        0         0
Bangalore      1         0
Mumbai         0         1
```

Here, `Chennai` is the reference category.

This is commonly called the **dummy variable trap** concept. If we keep all categories along with an intercept in certain statistical/linear models, the columns can become perfectly dependent on each other.

In Pandas, we can create dummy variables using `get_dummies()`:


In this example, `Bangalore` becomes the reference category because it is dropped first.

The interpretation is:

```text
City_Chennai = 1 → Chennai
City_Mumbai  = 1 → Mumbai
Both are 0       → Bangalore
```

So the dropped category is not lost. It is represented by the situation where all dummy columns are `0`.

Dummy Variables are especially useful in:

* Linear Regression
* Logistic Regression
* Statistical models
* Other models where avoiding redundant columns is important

One-Hot Encoding and Dummy Variables are closely related, but the common distinction is:

```text
One-Hot Encoding:
Create a column for every category.

Dummy Variables:
Create k - 1 columns for k categories,
using one category as the reference.
```

For example, with 3 categories:

```text
One-Hot → 3 columns
Dummy   → 2 columns
```

The key point is: **Dummy Variables represent categorical values using binary columns, usually keeping one category as the reference category to avoid unnecessary redundancy.**


# Frequency Encoding

Frequency Encoding is a technique that replaces each categorical value with the **number of times that category appears** in the dataset.

It is useful when a categorical column has many unique categories, because it can represent the category using a **single numerical column** instead of creating many columns.


First, we calculate how many times each city appears:

```text
Chennai   → 3
Bangalore → 2
Mumbai    → 1
```

Then we replace each category with its frequency:

```text
        City        Frequency
0    Chennai              3
1  Bangalore              2
2    Chennai              3
3     Mumbai              1
4    Chennai              3
5  Bangalore              2

Frequency Encoding can also use the **frequency proportion** instead of the raw count.

For example:

```text
Chennai   → 3/6 = 0.50
Bangalore → 2/6 = 0.33
Mumbai    → 1/6 = 0.17
```

This can be calculated using:

```python
frequency = df["City"].value_counts(normalize=True)

df["City_Frequency"] = df["City"].map(frequency)

print(df)
```

Frequency Encoding is useful when:

* A categorical feature has many unique values.
* One-Hot Encoding would create too many columns.
* The frequency of a category contains useful information.
* We want to keep the feature as a single numerical column.

For example, in an e-commerce dataset:

```text
Product_ID
P101
P102
P103
...
```

If there are thousands of products, One-Hot Encoding could create thousands of columns. Frequency Encoding can instead represent each product using one numerical value indicating how frequently it occurs.

However, Frequency Encoding has an important limitation.

Different categories can have the **same frequency**:

```text
Chennai   → 3
Delhi     → 3
Mumbai    → 2
```

The model cannot distinguish Chennai and Delhi based on the encoded value because both become `3`.

Also, frequency does not necessarily represent the relationship between the category and the target variable.

The key point is: **Frequency Encoding replaces each category with how frequently it occurs and is especially useful for high-cardinality categorical variables.**


````python

frequency = df["City"].value_counts()

df["City_Frequency"] = df["City"].map(frequency)

print(df)

````

# Target Encoding

Target Encoding is a technique that replaces each categorical value with a **statistical value calculated from the target variable**.

The most common approach is to replace each category with the **mean of the target variable for that category**.

For example, suppose we want to predict whether a customer will buy a product:


Now calculate the average target value for each city.

For Chennai:

```text
(1 + 1) / 2 = 1.00
```

For Bangalore:

```text
(0 + 1) / 2 = 0.50
```

For Mumbai:

```text
0 / 1 = 0.00
```

Therefore:

```text
Chennai   → 1.00
Bangalore → 0.50
Mumbai    → 0.00
```

We can implement this using Pandas:


The meaning is:

```text
Chennai   → 1.00 → higher purchase rate
Bangalore → 0.50 → medium purchase rate
Mumbai    → 0.00 → lower purchase rate
```

Target Encoding is useful when:

* A categorical variable has many unique categories.
* One-Hot Encoding would create too many columns.
* The category has a meaningful relationship with the target.
* We are working with supervised Machine Learning problems.

For example, consider a dataset containing thousands of `Product_ID` values. Instead of creating thousands of One-Hot columns, Target Encoding can represent each product using its historical average target value.

However, Target Encoding has a **major risk: target leakage**.

Suppose we calculate the target mean using the entire dataset before splitting it into training and testing data. Information from the target values can indirectly enter the features used for training.

This can make the model appear much better during evaluation than it actually is.

Another problem occurs with categories that have very few observations. For example:

```text
Product_A → 1 observation → target = 1
```

The encoded value becomes `1.0`, but that does not necessarily mean Product_A always produces a positive result. There is simply not enough data.

Therefore, Target Encoding should be performed carefully using techniques such as:

* Train/test separation
* Cross-validation
* Smoothing
* Out-of-fold encoding

The key point is: **Target Encoding uses information from the target variable to represent categories, making it powerful for high-cardinality data but vulnerable to target leakage and overfitting if used incorrectly.**


````python

import pandas as pd

df = pd.DataFrame({
    "City": ["Chennai", "Chennai", "Bangalore", "Bangalore", "Mumbai"],
    "Purchased": [1, 1, 0, 1, 0]
})

print(df)
````

# Handling Unknown Categories

Unknown categories occur when a categorical value appears during **validation, testing, or prediction** that was not present when the encoder was fitted on the training data.

For example, suppose the training data contains:

```text
City
------
Chennai
Bangalore
Mumbai
```

The encoder learns these categories:

```text
Chennai
Bangalore
Mumbai
```

Later, the test data contains:

```text
City
------
Chennai
Delhi
Mumbai
```

`Delhi` is an **unknown category** because it was not present in the training data.

If the encoder does not know how to handle `Delhi`, it may produce an error during transformation.

Using `OneHotEncoder`, we can handle unknown categories by using:

```python
handle_unknown="ignore"
```
The encoder learned:

```text
Bangalore
Chennai
Mumbai
```

When it sees `Delhi`, it does not create a new column because `Delhi` was not present during training.

The encoded test data will look conceptually like:

```text
Delhi → 0 0 0
```

The important point is that the **same columns learned from the training data are maintained**.

This is especially important in Machine Learning because new categories can appear in real-world data.

For example:

```text
Training Data:
Chennai
Bangalore
Mumbai

Future Data:
Chennai
Bangalore
Delhi
```

The model must be able to process `Delhi` without crashing.

Unknown categories can occur because of:

* New cities or countries
* New products
* New customers
* New departments
* New device types
* New categories introduced after model training

Different encoding techniques handle unknown categories differently. Therefore, the encoder should be configured properly before using it on validation, test, or production data.

A common best practice is:

```text
Training Data
      ↓
Fit Encoder
      ↓
Learn Categories
      ↓
Transform Training Data
      ↓
Transform Test/Future Data
      ↓
Handle Unknown Categories Safely
```

We should **never fit the encoder separately on the test data**, because this can cause inconsistent feature representations and may introduce data leakage.

The key point is: **Always prepare the encoder using training data and configure it to safely handle categories that may appear later but were not seen during training.**


````python


import pandas as pd
from sklearn.preprocessing import OneHotEncoder

train = pd.DataFrame({
    "City": ["Chennai", "Bangalore", "Mumbai"]
})

test = pd.DataFrame({
    "City": ["Chennai", "Delhi", "Mumbai"]
})

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

encoder.fit(train[["City"]])

train_encoded = encoder.transform(train[["City"]])
test_encoded = encoder.transform(test[["City"]])

print(test_encoded)
````

# High-Cardinality Categories

High-cardinality categories are categorical variables that contain a **large number of unique categories**.

For example:

```text id="2w3g8m"
Product_ID:
P001
P002
P003
...
P10000
```

Here, `Product_ID` has thousands of unique values, so it is a high-cardinality categorical variable.

Other examples include:

```text id="b9w7kx"
Customer_ID
Product_ID
Zip_Code
Email_ID
Website_URL
Transaction_ID
```

We can check the number of unique categories using:

If a dataset contains 100,000 different products, One-Hot Encoding could create approximately 100,000 columns for that single feature.

For example:

```text id="4l1r8h"
Product_ID
   ↓
One-Hot Encoding
   ↓
P001  P002  P003  P004 ... P100000
```

This can cause several problems:

* Very large number of features
* High memory usage
* Slower model training
* More complex models
* Sparse datasets
* Increased risk of overfitting

Therefore, One-Hot Encoding is often not the best choice for high-cardinality variables.

Instead, we can consider techniques such as:

```text id="xj1r8p"
High Cardinality
      ↓
Frequency Encoding
Target Encoding
Hash Encoding
Other domain-specific approaches
```

For example, Frequency Encoding can represent thousands of product IDs using only one numerical column:

```text id="h7u9k3"
Product_ID    Frequency
P001             500
P002             250
P003             100
```

Target Encoding can also represent each category using a statistic related to the target:

```text id="s6r0mj"
Product_ID    Target_Mean
P001             0.72
P002             0.45
P003             0.81
```

However, Target Encoding must be handled carefully because it can cause **target leakage and overfitting**.

Another important consideration is whether the categorical column should be treated as a feature at all.

For example:

```text id="j1a8r4"
Transaction_ID:
TX001
TX002
TX003
...
```

If every transaction ID is unique and has no meaningful relationship with the target, encoding it may not provide useful information. In such cases, the column may be better removed rather than encoded.

The key point is: **High-cardinality categories contain many unique values, so blindly applying One-Hot Encoding can create too many features. Choose an encoding technique based on the number of categories, the meaning of the feature, and its relationship with the target.**


````python

import pandas as pd

df = pd.DataFrame({
    "Product_ID": [
        "P001", "P002", "P003",
        "P004", "P005", "P006"
    ]
})

print(df["Product_ID"].nunique())
````
