# Why Use Pipelines?

## Explanation

A **Pipeline** in Scikit-learn is used to combine multiple data preprocessing steps into a single workflow.

For example, numerical data may need:

```text
Missing Value Imputation → Scaling
```

Categorical data may need:

```text
Missing Value Imputation → Encoding
```

Instead of performing each step separately, a Pipeline connects them in the correct order.

A typical machine learning workflow looks like:

```text
Raw Data
   ↓
Preprocessing
   ↓
Machine Learning Model
   ↓
Prediction
```

A Pipeline makes this workflow easier to manage and reuse.

## Why Is It Required?

Without a Pipeline, we need to manually perform every preprocessing step.

This can cause problems such as:

* Repeating the same code
* Applying steps in the wrong order
* Forgetting a preprocessing step
* Applying different preprocessing to training and test data
* Data leakage

A Pipeline ensures that the preprocessing steps are performed **in the same order and using the same rules**.

The missing value in `Age` is first replaced with the mean value.

Then `StandardScaler` scales the completed data.

The Pipeline automatically performs both operations in the correct order:

```text
Age
 ↓
Imputation
 ↓
Scaling
 ↓
Processed Data
```

Explain why the preprocessing technique is required and what problem it solves.

The Pipeline is required because real-world machine learning data often needs multiple preprocessing operations.

Instead of manually running:

```text
Imputation
↓
Scaling
↓
Encoding
↓
Model
```

we can connect the steps into one reusable workflow.

This reduces preprocessing errors and helps ensure that the same transformations are applied consistently to new data.

## Real-World Example

Imagine a hospital dataset containing patient information.

```text
Age
Weight
Blood Pressure
Gender
```

Some patients may have missing values.

Before using this data for prediction:

```text
Age, Weight, Blood Pressure
        ↓
Fill Missing Values
        ↓
Scale Numerical Data
```

A Pipeline can perform these steps automatically.

## Business Example

A bank wants to predict whether a customer will default on a loan.

Customer data may contain missing values and numerical features with different scales.

The bank can create a Pipeline:

```text
Customer Data
     ↓
Missing Value Imputation
     ↓
Feature Scaling
     ↓
Loan Default Model
```

This makes the preprocessing workflow consistent for both training data and new customer data.

## AI/ML Use Case

An AI/ML engineer building a **customer churn prediction model** may have features such as:

```text
Age
Monthly Charges
Tenure
Contract Type
Payment Method
```

The numerical features may require imputation and scaling, while categorical features require imputation and encoding.

Pipelines allow these preprocessing operations to be organized and later connected to the machine learning model.

```text
Raw Customer Data
       ↓
Preprocessing Pipeline
       ↓
Processed Features
       ↓
ML Model
       ↓
Churn Prediction
```

Therefore, Pipelines are important for building **clean, consistent, reusable, and reliable machine learning workflows**.


````python
## Python Example

import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "Age": [25, 30, None, 40]
})

pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler())
])

result = pipeline.fit_transform(df)

print(result)
````

# Pipeline

## Explanation

`Pipeline` in Scikit-learn is used to connect multiple preprocessing steps into a single sequence.

Each step is executed in a fixed order.

For example:

```text
Raw Data
   ↓
Imputation
   ↓
Scaling
   ↓
Processed Data
```

Instead of writing separate code for each operation, we can combine them using `Pipeline`.

## Why Is It Required?

`Pipeline` is required when multiple preprocessing steps need to be performed one after another.

It solves problems such as:

* Repeating preprocessing code
* Applying steps in the wrong order
* Forgetting a preprocessing step
* Inconsistent preprocessing between training and testing data

For example, scaling should happen **after** missing values have been handled.

`Pipeline` ensures the correct order automatically.


The Pipeline performs two operations automatically:

```text
Age
 ↓
SimpleImputer
 ↓
StandardScaler
 ↓
Processed Age
```

First, the missing value is replaced with the mean.

The values are then standardized by `StandardScaler`.

Explain why the preprocessing technique is required and what problem it solves.

`Pipeline` is required because preprocessing usually involves multiple dependent steps.

For example, scaling cannot properly handle missing values. Therefore, missing values should be handled first and scaling should happen afterward.

The Pipeline keeps these operations together and guarantees that they are performed in the defined order.

## Real-World Example

Consider a student dataset containing:

```text
Age
Marks
Attendance
```

Some values may be missing.

Before using the data for prediction:

```text
Student Data
     ↓
Fill Missing Values
     ↓
Scale Numerical Features
     ↓
Prediction
```

A Pipeline can perform these steps automatically.

## Business Example

A bank wants to predict loan approval.

Customer data may contain missing income values and numerical features with different scales.

The bank can create:

```text
Customer Data
     ↓
Imputation
     ↓
Scaling
     ↓
Loan Prediction Model

The Pipeline makes this process consistent for every customer.

## AI/ML Use Case

An AI/ML engineer building a house-price prediction model may have:

```text
House Size
Bedrooms

```

Some values may be missing, and the numerical features may have different ranges.

A Pipeline can handle both:

```text
House Data
    ↓

StandardScaler

Machine Learning Model
```

This makes the preprocessing workflow reusable during model training and prediction.


````python

## Python Example

import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "Age": [20, 30, None, 40, 50]
})

pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler())
])

result = pipeline.fit_transform(df)

print(result)
````

# ColumnTransformer

## Explanation

`ColumnTransformer` in Scikit-learn is used to apply **different preprocessing techniques to different columns** of a dataset.

This is useful because a dataset can contain different types of features.

For example:

```text
Numerical Features
Age, Salary
    ↓
Imputation → Scaling


Categorical Features
City, Gender
    ↓
Imputation → Encoding
```

`ColumnTransformer` allows us to combine these different preprocessing pipelines into one preprocessing workflow.

## Why Is It Required?

`ColumnTransformer` is required when numerical and categorical features need different preprocessing.

For example:

* Numerical columns need missing-value imputation and scaling.
* Categorical columns need missing-value imputation and encoding.

Applying the same preprocessing technique to every column would be incorrect.

`ColumnTransformer` solves this problem by sending each column to the appropriate preprocessing pipeline.


The exact numerical values come from `StandardScaler`, while the `0` and `1` values come from `OneHotEncoder`.

The processing happens like this:

```text
Age, Salary
    ↓
Numerical Pipeline
    ↓
Imputation → Scaling


City
    ↓
Categorical Pipeline
    ↓
Imputation → Encoding
```

`ColumnTransformer` combines the results from both pipelines into one processed dataset.

Explain why the preprocessing technique is required and what problem it solves.

A real-world dataset usually contains a mixture of numerical and categorical features.

For example:

```text
Age       → Numerical
Salary    → Numerical
City      → Categorical
Gender    → Categorical
```

These columns cannot all be processed in the same way.

`ColumnTransformer` solves this problem by allowing different preprocessing rules for different groups of columns.

## Real-World Example

Consider an employee dataset:

```text
Age
Salary
Department
City
```

The numerical features can be:

```text
Age, Salary
    ↓
Imputation
    ↓
Scaling
```

The categorical features can be:

```text
Department, City
    ↓
Imputation
    ↓
Encoding
```

`ColumnTransformer` combines both workflows into one preprocessing process.

## Business Example

A bank wants to predict whether a customer will receive a loan.

Customer information may contain:

```text
Age
Income
Loan Amount
Employment Type
City
```

The bank can process the data as:

```text
Age, Income, Loan Amount
        ↓
Imputation → Scaling

Employment Type, City
        ↓
Imputation → Encoding
```

`ColumnTransformer` ensures that each feature receives the correct preprocessing.

## AI/ML Use Case

An AI/ML engineer building a **loan approval prediction model** may use:

```text
Numerical:
Age
Income
Loan Amount

Categorical:
Employment Type
City
Education
```

The complete workflow can be:

```text
                 Raw Data
                    ↓
             ColumnTransformer
                ↙          ↘
        Numerical           Categorical
           ↓                    ↓
      Imputation            Imputation
           ↓                    ↓
        Scaling              Encoding
                ↘          ↙
              ML Model
```

Therefore, `ColumnTransformer` is important when a machine learning dataset contains **multiple feature types that require different preprocessing techniques**.


````python

## Python Example

import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

df = pd.DataFrame({
    "Age": [25, 30, None, 40],
    "Salary": [30000, 40000, 50000, None],
    "City": ["Chennai", "Bangalore", None, "Chennai"]
})

numerical_features = ["Age", "Salary"]
categorical_features = ["City"]

numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numerical", numerical_pipeline, numerical_features),
    ("categorical", categorical_pipeline, categorical_features)
])
result = preprocessor.fit_transform(df) 
print(result)
````

# Numerical Pipeline

## Explanation

A **Numerical Pipeline** is a sequence of preprocessing steps applied specifically to numerical features such as `Age`, `Salary`, `Income`, and `Experience`.

A common numerical pipeline is:

```text
Numerical Features
       ↓
Imputation
       ↓
Scaling
       ↓
Processed Numerical Features
```

For example, `SimpleImputer` can handle missing numerical values, and `StandardScaler` can scale the values.

## Why Is It Required?

Numerical data can have two common problems:

* Missing values
* Different numerical scales

For example:

```text
Age       → 25, 30, 40
Salary    → 30000, 50000, 70000
```

The values are on very different scales.

A numerical pipeline solves these problems by first filling missing values and then scaling the numerical features.


The pipeline performs two stages:

```text
Age, Salary
     ↓
Imputation
     ↓
Scaling
     ↓
Processed Numerical Data
```

The missing `Age` value is replaced with the mean age.

The missing `Salary` value is replaced with the mean salary.

After imputation, `StandardScaler` standardizes both numerical columns.

Explain why the preprocessing technique is required and what problem it solves.

The Numerical Pipeline is required because numerical features may contain missing values and may have very different scales.

**Imputation** prevents missing values from causing problems during model training.

**Scaling** puts numerical features on a comparable scale, which is important for algorithms that are sensitive to feature magnitude.

## Real-World Example

```text
House_Age

```

Some houses may have missing values.

The numerical pipeline can:

```text
House Age, Size, Price
          ↓
     Fill Missing Values

        Scaling

    Processed Features
```

This prepares the numerical data for a machine learning model.

```text

```

Some values may be missing, and income or loan amount may have much larger values than age.

```text

# Imputation Pipeline

An **Imputation Pipeline** is a pipeline used to handle missing values in a dataset before passing the data to a machine learning model.

Instead of manually filling missing values, we can use `Pipeline` to automatically perform imputation as part of the ML preprocessing workflow.

## Why is Imputation Pipeline Required?

Real-world datasets often contain missing values.

For example:

```text
Age
25
30
NaN
40
```

An ML model may not be able to work directly with `NaN` values.

An imputation pipeline can replace the missing value with a suitable value such as the mean or median.

## What the Python Example Tells

The pipeline contains one step:

```text
SimpleImputer
     ↓
Find missing value
     ↓
Calculate median
     ↓
Replace missing value
```

The values are:

```text
25
30
NaN
40
35
```

The median is `35`, so the missing value is replaced with `35`.

The result becomes:

```text
25
30
35
40
35
```

## AI/ML Example

Suppose a customer dataset contains:

```text
Age     Salary
25      30000
30      40000
NaN     35000
40      NaN
```

The imputation pipeline can fill the missing values before the data is passed to the ML model.

## Business Example

A bank's customer dataset may have missing `Annual_Income` values.

Instead of deleting those customers, the preprocessing pipeline can replace missing income values using a suitable strategy such as median imputation.

## AI/ML Engineer Use Case

An imputation pipeline is useful when building reusable ML preprocessing workflows where missing-value handling must happen automatically and consistently during model training and prediction.


````python

## Python Example

import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

df = pd.DataFrame({
    "Age": [25, 30, None, 40, 35]
})

pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

result = pipeline.fit_transform(df[["Age"]])

print(result)
````

# Scaling Pipeline

A **Scaling Pipeline** is a pipeline that applies a feature scaling technique to numerical columns before passing the data to a machine learning model.

Scaling puts numerical features into a comparable range so that features with larger values do not unnecessarily dominate features with smaller values.

## Why is Scaling Pipeline Required?

Consider:

```text
Age     Salary
25      30000
40      80000
```

`Salary` has much larger numerical values than `Age`.

For algorithms such as Logistic Regression, KNN, SVM, and K-Means, scaling can improve model performance.

A scaling pipeline combines the scaling step with the ML workflow.


## What the Python Example Tells

The pipeline contains `StandardScaler`.

The pipeline performs:

```text
Original Numerical Data
        ↓
StandardScaler
        ↓
Scaled Numerical Data
```

`StandardScaler` transforms each numerical column so that its values are centered around `0` with a standard deviation close to `1`.

For example:

```text
Age
20 → -1.34
30 → -0.45
40 →  0.45
50 →  1.34
```

The same scaling process is applied to `Salary`.

## AI/ML Example

For a customer classification model:

```text
Age
Income
Account_Balance
Transaction_Count
```

These features may have very different scales.

A scaling pipeline standardizes them before they are given to the model.

## Business Example

A bank is building a loan approval model using:

```text
Age              → 20–70
Annual Income    → 20,000–2,00,000
Credit Score     → 300–850
```

Scaling makes these numerical features comparable before training a scale-sensitive model.

## AI/ML Engineer Use Case

A scaling pipeline is useful when preprocessing numerical features for algorithms that are sensitive to feature magnitude. It also keeps preprocessing organized and reusable during both training and prediction.


````python

## Python Example

import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "Age": [20, 30, 40, 50],
    "Salary": [20000, 40000, 60000, 80000]
})

pipeline = Pipeline([
    ("scaler", StandardScaler())
])

result = pipeline.fit_transform(df)

print(result)
````

# Combining Multiple Transformations

In real-world datasets, different columns often require different preprocessing operations.

For example:

* Numerical columns may need **imputation + scaling**.
* Categorical columns may need **imputation + encoding**.

`ColumnTransformer` is used to apply different preprocessing pipelines to different groups of columns.

## Why Combine Multiple Transformations?

Suppose a customer dataset contains:

```text
Age       → Numerical
Salary    → Numerical
City      → Categorical
Gender    → Categorical
```

We should not apply the same transformation to every column.

Instead:

```text
Numerical Columns
       ↓
Imputation
       ↓
Scaling

Categorical Columns
       ↓
Imputation
       ↓
Encoding
```

`ColumnTransformer` combines these different transformations into one preprocessing workflow.


## What the Python Example Tells

The preprocessing is divided into two pipelines.

Numerical columns:

```text id="9v3d6a"
Age, Salary
      ↓
Median Imputation
      ↓
StandardScaler
      ↓
Scaled Numerical Features
```

Categorical column:

```text id="v5s2kd"
City
 ↓
Most-Frequent Imputation
 ↓
One-Hot Encoding
 ↓
Encoded Categorical Features
```

`ColumnTransformer` combines the results from both pipelines into one transformed dataset.

The final output contains:

```text id="3i5q9r"
Scaled Age
Scaled Salary
Encoded City
```

## AI/ML Example

For a customer churn model:

```text id="z4q8hx"
Age, Salary
     ↓
Imputation → Scaling

City, Gender, Plan
     ↓
Imputation → Encoding

          ↓
   Combined Features
          ↓
      ML Model
```

This allows the model to receive all required features in a single numerical dataset.

## Business Example

A bank's loan dataset may contain:

```text id="c7n2rx"
Age              → Numerical
Annual_Income    → Numerical
Credit_Score     → Numerical
City             → Categorical
Employment_Type  → Categorical
```

Numerical features can be imputed and scaled, while categorical features can be imputed and encoded.

## AI/ML Engineer Use Case

Combining multiple transformations is a standard approach for production ML preprocessing because different feature types often require different preprocessing operations. `ColumnTransformer` makes it possible to manage these transformations together in one consistent workflow.


````python

## Python Example

import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

df = pd.DataFrame({
    "Age": [25, 30, None, 40],
    "Salary": [30000, None, 50000, 60000],
    "City": ["Chennai", "Bangalore", None, "Mumbai"]
})

numeric_features = ["Age", "Salary"]
categorical_features = ["City"]

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

result = preprocessor.fit_transform(df)

print(result)
````

# fit()

`fit()` is used to **learn the required information from the training data**.

It does not transform the data. Instead, it calculates and stores the values needed for later transformation.

## How fit() Works

For different preprocessing techniques, `fit()` learns different information.

```text id="f1r8cz"
StandardScaler
     ↓
Learns mean and standard deviation

SimpleImputer
     ↓
Learns median / mean / most frequent value

OneHotEncoder
     ↓
Learns unique categories

## What the Python Example Tells

The `fit()` method looks at the `Age` values:

```text id="x3w9kp"
20
30
40
50
```

It calculates:

```text id="k7m2fd"
Mean = 35
Standard Deviation ≈ 11.18
```

These values are stored inside the `StandardScaler`.

The data has **not been scaled yet**.

To actually scale the data, we use `transform()`.

```text id="q6v8nz"
fit()
  ↓
Learn required values
  ↓
Store them
```

## AI/ML Example

During ML training:

```text id="n5r2wt"
Training Data
     ↓
fit()
     ↓
Learn preprocessing parameters
     ↓
Store parameters
```

For example, `StandardScaler` learns the mean and standard deviation from the training data.

## Business Example

A bank is building a loan prediction model.

If `Annual_Income` is scaled, `fit()` calculates the required scaling parameters from the **training customers**.

These learned parameters are then used to process future customer data.

## AI/ML Engineer Use Case

`fit()` should generally be performed on the **training data only**. This prevents information from the test data from influencing the preprocessing process and helps avoid data leakage.


````python


## Python Example

import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "Age": [20, 30, 40, 50]
})

scaler = StandardScaler()

scaler.fit(df[["Age"]])

print("Mean:", scaler.mean_)
print("Standard Deviation:", scaler.scale_)
````

# transform()

`transform()` is used to **apply the information learned by `fit()` to data**.

It does not learn new parameters. It uses the parameters that were already learned during `fit()`.

## How transform() Works

The process is:

```text
Training Data
     ↓
fit()
     ↓
Learn Parameters
     ↓
transform()
     ↓
Transformed Data
```

For example, `StandardScaler` learns the mean and standard deviation using `fit()`. Then `transform()` uses those values to scale the data.

## What the Python Example Tells

First, `fit()` learns the required information:

```text id="r8c3mv"
Mean = 35
Standard Deviation ≈ 11.18
```

Then `transform()` uses those learned values to scale each age.

For example:

```text id="b2k9rx"
20 → -1.34
30 → -0.45
40 →  0.45
50 →  1.34
```

So:

```text id="v4j7pd"
fit()
  ↓
Learn parameters

transform()
  ↓
Apply parameters
```

## AI/ML Example

During ML training:

```text id="e5m8ws"
Training Data
     ↓
fit()
     ↓
Learn scaling parameters
     ↓
transform()
     ↓
Scaled Training Data
```

For test data, we use the **same learned parameters**:

```text id="a9q4zt"
Test Data
    ↓
transform()
    ↓
Scaled Test Data
```

We do **not** call `fit()` on the test data.

## Business Example

A bank trains a loan prediction model using customer income data.

The scaler learns the income statistics from the training customers. When a new customer applies for a loan, `transform()` uses those same statistics to scale the new customer's income.

## AI/ML Engineer Use Case

`transform()` is important for maintaining consistent preprocessing between training, testing, and production data. It ensures that new data is processed using the parameters learned from the training data.


````python

## Python Example

import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "Age": [20, 30, 40, 50]
})

scaler = StandardScaler()

scaler.fit(df[["Age"]])

result = scaler.transform(df[["Age"]])

print(result)
````

# fit_transform()

`fit_transform()` is a combination of **`fit()` and `transform()`**.

It first learns the required parameters from the data and then immediately transforms the same data using those learned parameters.

## How fit_transform() Works

```text id="q8v4mx"
Data
 ↓
fit()
 ↓
Learn Parameters
 ↓
transform()
 ↓
Transformed Data
```

`fit_transform()` performs both steps together.


## What the Python Example Tells

This line:

```python id="n7c3wp"
result = scaler.fit_transform(df[["Age"]])
```

does two operations:

```text id="d4r8yk"
fit()
 ↓
Learn mean and standard deviation
 ↓
transform()
 ↓
Scale the Age values
```

The input:

```text id="v6k2mn"
20
30
40
50
```

is converted into:

```text id="j9p4sx"
-1.34
-0.45
 0.45
 1.34
```

## fit() vs transform() vs fit_transform()

```text id="a5w8qe"
fit()
→ Learn parameters

transform()
→ Apply learned parameters

fit_transform()
→ Learn parameters + Apply transformation
```

## AI/ML Example

For training data, we commonly use:

```python id="u3m6pk"
X_train = scaler.fit_transform(X_train)
```

For test data, we use:

```python id="r8n2vz"
X_test = scaler.transform(X_test)
```

This is important because the scaler should learn its parameters **only from the training data**.

## Business Example

A bank has training customer data and test customer data.

```text id="e7q4ws"
Training Data
     ↓
fit_transform()
     ↓
Learn + Transform

Test Data
     ↓
transform()
     ↓
Transform using training parameters
```

This prevents information from the test data from leaking into the training process.

## AI/ML Engineer Use Case

`fit_transform()` is commonly used on training data because the preprocessing parameters need to be learned from the training data and applied immediately. For validation, test, and production data, use `transform()` with the already-fitted preprocessing object.


````python
## Python Example

import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "Age": [20, 30, 40, 50]
})

scaler = StandardScaler()

result = scaler.fit_transform(df[["Age"]])

print(result)
````

# Complete Preprocessing Pipeline

A **complete preprocessing pipeline** combines all required preprocessing steps for numerical and categorical features into one workflow.

For this example:

```text
Numerical Features
        ↓
   Imputation
        ↓
     Scaling

Categorical Features
        ↓
   Imputation
        ↓
    Encoding
```

`ColumnTransformer` combines both pipelines into one preprocessing object.

## Why Use a Complete Preprocessing Pipeline?

A real-world dataset usually contains both numerical and categorical columns.

For example:

```text
Age              → Numerical
Salary           → Numerical
City             → Categorical
Gender           → Categorical
```

Each type requires different preprocessing.

A complete pipeline ensures that the correct preprocessing is applied to each column.


## What the Python Example Tells

The complete preprocessing happens in stages.

### Stage 1: Select Numerical Features

```text id="p6k3rt"
Age
Salary
```

These columns are sent to the numerical pipeline.

### Stage 2: Numerical Imputation

```text id="h2v7qx"
Age       → Missing value
Salary    → Missing value
```

`SimpleImputer(strategy="median")` replaces missing numerical values with the median of the respective column.

```text id="s5m8wn"
Numerical Data
      ↓
Median Imputation
```

### Stage 3: Numerical Scaling

After missing values are handled:

```text id="c9r4yk"
Imputed Numerical Data
        ↓
StandardScaler
        ↓
Scaled Numerical Data
```

The numerical values are standardized.

### Stage 4: Select Categorical Features

```text id="x7p2mv"
City
Gender
```

These columns are sent to the categorical pipeline.

### Stage 5: Categorical Imputation

Missing categorical values are replaced using the most frequent category.

```text id="w3k8qa"
Categorical Data
       ↓
Most-Frequent Imputation
```

### Stage 6: Categorical Encoding

After missing values are handled:

```text id="r6n4tz"
Imputed Categorical Data
        ↓
OneHotEncoder
        ↓
Numerical Encoded Data
```

The categorical values are converted into numerical columns.

### Stage 7: Combine Everything

`ColumnTransformer` combines the processed numerical and categorical features:

```text id="y8q2mk"
Numerical Pipeline
Imputation → Scaling
        ↓
        ├──────────────┐
                       ↓
                Combined Data
                       ↑
        ┌──────────────┘
        ↓
Categorical Pipeline
Imputation → Encoding
```

The final output contains only numerical values, which can be passed to an ML model.

## AI/ML Example

A customer churn model may use:

```text id="a4w9px"
Age, Salary
    ↓
Imputation → Scaling

City, Gender
    ↓
Imputation → Encoding

        ↓
Processed Features
        ↓
Churn Prediction Model
```

## Business Example

A bank wants to predict whether a customer will default on a loan.

The dataset contains:

```text id="m5v7kc"
Age              → Numerical
Annual_Income    → Numerical
Credit_Score     → Numerical
City             → Categorical
Employment_Type  → Categorical
```

The complete preprocessing pipeline automatically handles missing values, scales numerical features, encodes categorical features, and produces data ready for the ML model.

## AI/ML Engineer Use Case

This is a common production ML pattern because the entire preprocessing workflow can be kept together and later combined with a machine learning model.

```text id="z3n8qw"
Raw Data
   ↓
ColumnTransformer
   ↓
Numerical: Imputation → Scaling
Categorical: Imputation → Encoding
   ↓
Processed Data
   ↓
ML Model
```
