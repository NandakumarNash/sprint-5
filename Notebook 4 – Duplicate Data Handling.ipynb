# What are Duplicate Records?

Duplicate records are rows in a dataset that represent the **same data or the same real-world entity more than once**.

Duplicates can happen because of:

* Repeated data entry
* Multiple file uploads
* System errors
* Data merging from different sources
* Repeated customer transactions or registrations

### Simple Example

```text
Customer_ID   Name    Age
101           Arun    25
102           Ravi    30
101           Arun    25
```

The first and third rows are duplicates because they contain exactly the same information.

### Real-World Example

A customer registers twice on an e-commerce website because of a technical issue.

```text
Customer_ID   Name      Email
101           Arun      arun@gmail.com
101           Arun      arun@gmail.com
```

### Business Example

If a sales record is duplicated, the company may calculate:

```text
Actual Sales = ₹50,000
Recorded Sales = ₹60,000
```

This can lead to incorrect business reports.

### AI/ML Use Case

Duplicate records can cause some observations to appear more frequently than they should.

For example, if a customer appears 10 times instead of once, an ML model may give that customer group **too much importance**.

Therefore, duplicate detection is an important part of data preprocessing.


# Exact Duplicates

An **exact duplicate** occurs when two or more rows contain exactly the same values in every column.

### Example

```text
Customer_ID   Name    Age   City
101           Arun    25    Chennai
102           Ravi    30    Bangalore
101           Arun    25    Chennai
```

Here, rows 1 and 3 are exact duplicates.

### Python Example

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 102, 101],
    "Name": ["Arun", "Ravi", "Arun"],
    "Age": [25, 30, 25],
    "City": ["Chennai", "Bangalore", "Chennai"]
})

print(df.duplicated())
```

Output:

```text
0    False
1    False
2     True
dtype: bool
```

`True` means the row is an exact duplicate of a previous row.

### Real-World Example

A customer transaction is accidentally inserted twice:

```text
Order_ID   Product   Amount
5001       Laptop    50000
5001       Laptop    50000
```

The second row is an exact duplicate.

### AI/ML Use Case

Exact duplicates can cause the same observation to be counted multiple times during model training, potentially creating bias in the dataset.


# Partial Duplicates

A **partial duplicate** occurs when records represent the same entity but are not completely identical.

Some columns may be different.

### Example

```text
Customer_ID   Name    Email              City
101           Arun    arun@gmail.com     Chennai
101           Arun    arun@gmail.com     Bangalore
```

These rows are not exact duplicates because the `City` values are different.

However, they may represent the **same customer**.

### Why Partial Duplicates Happen

* Customer information was updated
* Different systems contain different information
* Data entry mistakes
* Multiple registrations
* Data integration from different sources

### Business Example

A customer changes their city:

```text
Customer_ID   Name    City
101           Arun    Chennai
101           Arun    Bangalore
```

Deleting one row blindly could remove valid information.

### AI/ML Use Case

Suppose a customer appears multiple times with different information. If these records are treated as separate customers, an ML model may learn incorrect patterns.

Therefore, partial duplicates usually require **business rules** before removal.


# Identifying Duplicates

Before removing duplicates, we should first **identify how many duplicate records exist and which records are duplicated**.

### Python Example

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 102, 101, 103],
    "Name": ["Arun", "Ravi", "Arun", "Priya"],
    "Age": [25, 30, 25, 28]
})

print(df.duplicated())
```

Output:

```text
0    False
1    False
2     True
3    False
```

### Find Number of Duplicates

```python
print(df.duplicated().sum())
```

Output:

```text
1
```

This means there is **1 duplicate row**.

### Display Duplicate Rows

```python
print(df[df.duplicated()])
```

Output:

```text
   Customer_ID  Name  Age
2          101  Arun   25
```

### Important Point

Always **identify and understand duplicates before deleting them**.

A duplicate-looking record may sometimes contain valid business information.


# `duplicated()`

The `duplicated()` method is used to **identify duplicate rows** in a Pandas DataFrame.

### Syntax

```python
df.duplicated()
```

It returns a Boolean value:

```text
False → Not a duplicate
True  → Duplicate
```

### Example

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Ravi", "Arun"],
    "Age": [25, 30, 25]
})

print(df.duplicated())
```

Output:

```text
0    False
1    False
2     True
dtype: bool
```

### Count Duplicates

```python
print(df.duplicated().sum())
```

Output:

```text
1
```

### Display Duplicate Rows

```python
print(df[df.duplicated()])
```

Output:

```text
   Name  Age
2  Arun   25
```

By default, `duplicated()` considers the **first occurrence as original** and marks later occurrences as duplicates.

### Keep Last Occurrence

```python
df.duplicated(keep="last")
```

### Mark All Copies

```python
df.duplicated(keep=False)
```

`keep=False` marks **all occurrences** of duplicated records as `True`.


# `drop_duplicates()`

The `drop_duplicates()` method is used to **remove duplicate rows** from a DataFrame.

### Example

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Ravi", "Arun"],
    "Age": [25, 30, 25]
})

df = df.drop_duplicates()

print(df)
```

Output:

```text
   Name  Age
0  Arun   25
1  Ravi   30
```

The second occurrence of Arun was removed.

### Keep Last Record

```python
df.drop_duplicates(keep="last")
```

This keeps the last occurrence instead of the first.

### Remove All Duplicate Groups

```python
df.drop_duplicates(keep=False)
```

This removes **all records that have duplicates**.

### Important

`drop_duplicates()` returns a cleaned DataFrame.

You can save it using:

```python
df = df.drop_duplicates()
```

or:

```python
df.drop_duplicates(inplace=True)
```

### Business Example

If the same order was accidentally inserted twice:

```text
Order_ID   Amount
1001       5000
1001       5000
```

Removing the duplicate prevents the order from being counted twice.

However, **do not automatically remove every duplicate without understanding the business meaning of the records**.


# Duplicate Detection Based on Selected Columns

Sometimes we don't want to compare the **entire row**.

We may want to identify duplicates based on specific columns.

This is especially useful for **partial duplicates**.

### Example

```text
Customer_ID   Name    Email              City
101           Arun    arun@gmail.com     Chennai
101           Arun    arun@gmail.com     Bangalore
102           Ravi    ravi@gmail.com     Chennai
```

Here, the first two rows have different cities but the same:

```text
Customer_ID
Name
Email
```

### Python Example

```python
print(df.duplicated(subset=["Customer_ID", "Email"]))
```

Output:

```text
0    False
1     True
2    False
dtype: bool
```

The second row is considered a duplicate based on the selected columns.

### Remove Based on Selected Columns

```python
df = df.drop_duplicates(
    subset=["Customer_ID", "Email"]
)
```

### Business Example

A company may define:

```text
Customer_ID + Email = Unique Customer
```

If the same combination appears multiple times, it may indicate duplicate customer records.

### AI/ML Use Case

Before training a model, duplicate customers or transactions can be detected using business-specific identifiers instead of comparing every column.
